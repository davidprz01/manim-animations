## 1. ¿Qué es Manim?

**Manim** (Mathematical Animation Engine) es una biblioteca de Python para crear animaciones matemáticas precisas y de alta calidad.

### Historia:
- **Manim Original (3b1b)**: Creado por Grant Sanderson para su canal de YouTube [3Blue1Brown](https://www.youtube.com/c/3blue1brown)
- **Manim Community**: Fork mantenido por la comunidad con mejor documentación y estabilidad

### ¿Por qué usar Manim?
- ✅ Animaciones matemáticas profesionales
- ✅ Control preciso de cada elemento visual
- ✅ Renderizado de alta calidad (hasta 4K)
- ✅ Código Python = reproducibilidad total
- ✅ Gran comunidad y documentación

### Versiones de Manim:
| Versión | Descripción | Recomendado |
|---------|-------------|-------------|
| **Manim Community** | Fork estable y documentado | ✅ Sí |
| Manim 3b1b | Original de Grant Sanderson | Para usuarios avanzados |
| ManimGL | Versión OpenGL (tiempo real) | Experimental |

> 📌 **Este curso usa Manim Community v0.19.0**

## 4. Configuración del Entorno

### VS Code (Recomendado)

**Extensiones útiles:**

- **Jupyter** - Notebooks interactivos
- **Manim Sideview** - Preview en tiempo real

### Jupyter Notebooks

Para usar Manim en Jupyter Notebooks, usamos el **magic command**:

```python
%%manim -qm -v WARNING NombreClase
```

**Parámetros del magic:**
| Parámetro | Descripción |
|-----------|-------------|
| `-ql` | Calidad baja (480p, 15fps) |
| `-qm` | Calidad media (720p, 30fps) |
| `-qh` | Calidad alta (1080p, 60fps) |
| `-qk` | Calidad 4K (2160p, 60fps) |
| `-v WARNING` | Nivel de verbosidad |
| `-p` | Preview automático |
| `-s` | Solo última imagen |


## 6. Tu Primera Animación

### Anatomía de una Escena Manim:

```python
from manim import *  # Importar todo de manim

class MiEscena(Scene):  # Heredar de Scene
    def construct(self):  # Método principal
        # 1. Crear objetos (Mobjects)
        texto = Text("¡Hola Manim!")
        
        # 2. Añadir/Animar objetos
        self.play(Write(texto))
        
        # 3. Esperar
        self.wait(2)
```

### Conceptos Clave:
- **Scene**: Contenedor de la animación
- **Mobject**: Objeto matemático (Mathematical Object)
- **Animation**: Transición visual de un Mobject
- **self.play()**: Ejecuta una animación
- **self.wait()**: Pausa en la animación
- **self.add()**: Añade objeto sin animación

````python
from manim import *
import numpy as np
````

````python
%%manim -qm -v WARNING PrimeraAnimacion

class PrimeraAnimacion(Scene):
    def construct(self):
        # Crear un texto de bienvenida
        titulo = Text("¡Bienvenido a Manim!", font_size=48, color=BLUE)
        subtitulo = Text("Tu primera animación", font_size=32, color=YELLOW)
        subtitulo.next_to(titulo, DOWN, buff=0.5)
        
        # Crear formas básicas
        circulo = Circle(radius=1, color=RED, fill_opacity=0.5)
        cuadrado = Square(side_length=2, color=GREEN, fill_opacity=0.5)
        triangulo = Triangle(color=PURPLE, fill_opacity=0.5)
        
        formas = VGroup(circulo, cuadrado, triangulo)
        formas.arrange(RIGHT, buff=1)
        formas.shift(DOWN)
        
        # Animaciones
        self.play(Write(titulo))
        self.play(FadeIn(subtitulo))
        self.wait(1)
        
        self.play(
            FadeOut(titulo),
            FadeOut(subtitulo)
        )
        
        # Mostrar formas una por una
        self.play(Create(circulo))
        self.play(Create(cuadrado))
        self.play(Create(triangulo))
        self.wait(1)
        
        # Transformaciones
        self.play(
            circulo.animate.shift(UP * 2),
            cuadrado.animate.rotate(PI/4),
            triangulo.animate.scale(1.5)
        )
        self.wait(1)
        
        # Final
        final_text = Text("¡Manim instalado correctamente!", color=GREEN, font_size=36)
        self.play(
            FadeOut(formas),
            Write(final_text)
        )
        self.wait(2)
````

## 7. Parámetros de Renderizado

### Desde Terminal:
```bash
manim -pql archivo.py NombreEscena    # Preview + baja calidad
manim -qh archivo.py NombreEscena     # Alta calidad
manim -qk archivo.py NombreEscena     # 4K
manim -a archivo.py                    # Todas las escenas
```

### Parámetros Comunes:
| Flag | Descripción |
|------|-------------|
| `-p` | Abrir preview al terminar |
| `-f` | Mostrar archivo en explorador |
| `-a` | Renderizar todas las escenas |
| `-s` | Solo guardar última imagen |
| `-n <num>` | Empezar desde frame N |
| `--format gif` | Exportar como GIF |
| `-o <nombre>` | Nombre del archivo de salida |

### Ejemplos:
```bash
# GIF de alta calidad
manim --format gif -qh archivo.py MiEscena

# Video con nombre personalizado
manim -qh -o "mi_video" archivo.py MiEscena

# Solo frames 10-50
manim -qm -n 10,50 archivo.py MiEscena
```

## 8. Configuración de Calidad

### Presets de Calidad:

| Preset | Resolución | FPS | Uso |
|--------|------------|-----|-----|
| `-ql` (low) | 854×480 | 15 | Desarrollo rápido |
| `-qm` (medium) | 1280×720 | 30 | Preview |
| `-qh` (high) | 1920×1080 | 60 | Producción |
| `-qp` (production) | 1920×1080 | 60 | Producción |
| `-qk` (fourk) | 3840×2160 | 60 | 4K |

### Configuración Personalizada:
```python
config.pixel_height = 1080
config.pixel_width = 1920
config.frame_rate = 60
config.background_color = WHITE
```

````python
%%manim -qm -v WARNING ComparacionCalidades

class ComparacionCalidades(Scene):
    def construct(self):
        titulo = Text("Configuración de Manim", font_size=40, color=BLUE)
        titulo.to_edge(UP)
        self.play(Write(titulo))
        
        # Tabla de calidades
        tabla_data = [
            ["Calidad", "Resolución", "FPS"],
            ["-ql (low)", "854×480", "15"],
            ["-qm (medium)", "1280×720", "30"],
            ["-qh (high)", "1920×1080", "60"],
            ["-qk (4K)", "3840×2160", "60"]
        ]
        
        tabla = Table(
            tabla_data,
            include_outer_lines=True,
            line_config={"stroke_width": 2}
        ).scale(0.6)
        
        # Colorear encabezados
        tabla.get_rows()[0].set_color(YELLOW)
        
        self.play(Create(tabla), run_time=2)
        self.wait(2)
        
        # Resaltar calidad recomendada para desarrollo
        rect = SurroundingRectangle(tabla.get_rows()[2], color=GREEN, buff=0.1)
        texto_rec = Text("← Recomendado para desarrollo", font_size=20, color=GREEN)
        texto_rec.next_to(rect, RIGHT)
        
        self.play(Create(rect), Write(texto_rec))
        self.wait(2)
````

## 9. Solución de Problemas Comunes

### Error: "manim not found"
```bash
# Verificar instalación
pip show manim

# Reinstalar
pip uninstall manim
pip install manim
```

### Error: "LaTeX not found"
- Instalar MiKTeX (Windows) o TeX Live (Linux/Mac)
- O usar `Text()` en lugar de `Tex()` / `MathTex()`

### Error: "FFmpeg not found"
```bash
# Windows (con Chocolatey)
choco install ffmpeg

# macOS
brew install ffmpeg

# Linux
sudo apt install ffmpeg
```

### Error: "No module named 'manim'"
```bash
# Verificar que estás en el entorno correcto
which python  # Linux/Mac
where python  # Windows

# Instalar en el entorno actual
pip install manim
```

### Video no se genera:
- Verificar que la clase hereda de `Scene`
- Verificar que el método se llama `construct`
- Verificar que hay al menos un `self.play()` o `self.wait()`

### Animación muy lenta:
- Usar `-ql` para desarrollo
- Reducir `run_time` de las animaciones
- Simplificar objetos complejos

## 11. Verificación Final

Ejecuta la siguiente celda para verificar que todo está configurado correctamente:

````python
%%manim -qm -v WARNING VerificacionInstalacion

class VerificacionInstalacion(Scene):
    def construct(self):
        # Título
        titulo = Text("✓ Verificación de Instalación", font_size=40, color=GREEN)
        titulo.to_edge(UP)
        
        # Lista de verificaciones
        checks = VGroup(
            Text("✓ Manim importado correctamente", font_size=28, color=GREEN),
            Text("✓ Renderizado de texto funcionando", font_size=28, color=GREEN),
            Text("✓ Formas básicas disponibles", font_size=28, color=GREEN),
            Text("✓ Animaciones ejecutándose", font_size=28, color=GREEN),
            Text("✓ FFmpeg funcionando", font_size=28, color=GREEN)
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.4)
        checks.next_to(titulo, DOWN, buff=1)
        
        # Formas de demostración
        demo = VGroup(
            Circle(radius=0.4, color=RED, fill_opacity=0.7),
            Square(side_length=0.8, color=BLUE, fill_opacity=0.7),
            Triangle(color=YELLOW, fill_opacity=0.7).scale(0.5)
        ).arrange(RIGHT, buff=0.5)
        demo.to_edge(DOWN, buff=1)
        
        # Animaciones
        self.play(Write(titulo))
        
        for check in checks:
            self.play(FadeIn(check, shift=RIGHT), run_time=0.5)
        
        self.wait(0.5)
        self.play(
            LaggedStart(*[GrowFromCenter(shape) for shape in demo], lag_ratio=0.3)
        )
        
        # Rotación de demostración
        self.play(
            Rotate(demo[0], angle=TAU),
            demo[1].animate.rotate(PI/4),
            demo[2].animate.scale(1.5),
            run_time=1.5
        )
        
        # Mensaje final
        final = Text("¡Todo listo para empezar!", font_size=36, color=GOLD)
        final.next_to(demo, UP, buff=0.5)
        self.play(Write(final))
        
        self.wait(2)
````
