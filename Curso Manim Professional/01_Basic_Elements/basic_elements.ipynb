## Estructura básica

```python
%%manim -qm -v Warning BasicStructure

class BasicStructure(Scene):
    def construct(self):
        # Toda escena tiene un método construct()
        # Aquí defines qué objetos aparecen y cómo se animan
        
        circle = Circle()
        self.play(Create(circle))
        self.wait(1)
```

## Mobjects

```python
%%manim -qm -v Warning IntroduccionMobjects

class IntroduccionMobjects(Scene):
    def construct(self):
        # Mobjects = Mathematical Objects
        # Son los objetos que se pueden animar
        
        # Formas básicas
        circulo = Circle(color=BLUE)
        cuadrado = Square(color=RED)
        triangulo = Triangle(color=GREEN)
        
        # Agrupar objetos
        formas = VGroup(circulo, cuadrado, triangulo)
        formas.arrange(RIGHT, buff=1)
        
        self.play(Create(formas))
        self.wait(2)
```

## Animaciones básicas

### Add vs Play

Hay dos formas principales de agregar objetos a la escena:
- `self.add()` - Aparece instantáneamente
- `self.play()` - Se anima la aparición

```python
%%manim -qm -v Warning AddVsPlay

class AddVsPlay(Scene):
    def construct(self):
        # Usando add() - aparece instantáneamente
        circulo = Circle(color=BLUE)
        self.add(circulo)
        self.wait(1)
        
        # Usando play() - se anima
        cuadrado = Square(color=RED).shift(RIGHT * 3)
        self.play(Create(cuadrado))
        self.wait(1)
```

### Animaciones de creación

```python
%%manim -qm -v Warning AnimacionesCreacion

class AnimacionesCreacion(Scene):
    def construct(self):
        # Create - dibuja el contorno
        circulo = Circle(color=BLUE)
        self.play(Create(circulo))
        self.wait(0.5)
        self.play(FadeOut(circulo))
        
        # Write - para texto
        texto = Text("¡Hola Manim!", color=YELLOW)
        self.play(Write(texto))
        self.wait(0.5)
        self.play(FadeOut(texto))
        
        # FadeIn - aparece gradualmente
        cuadrado = Square(color=RED)
        self.play(FadeIn(cuadrado))
        self.wait(0.5)
        self.play(FadeOut(cuadrado))
        
        # GrowFromCenter - crece desde el centro
        estrella = Star(color=GOLD)
        self.play(GrowFromCenter(estrella))
        self.wait(1)
```

### Wait - Pausas en la animación

```python
%%manim -qm -v Warning UsoDeWait

class UsoDeWait(Scene):
    def construct(self):
        circulo = Circle(color=BLUE)
        
        # Crear el círculo
        self.play(Create(circulo))
        
        # Esperar 2 segundos
        self.wait(2)
        
        # Cambiar color
        self.play(circulo.animate.set_color(RED))
        
        # Esperar 1 segundo
        self.wait(1)
```

## Ejemplo completo: Combinando conceptos

```python
%%manim -qm -v Warning EjemploCompleto

class EjemploCompleto(Scene):
    def construct(self):
        # Título
        titulo = Text("Elementos Básicos de Manim", font_size=40, color=BLUE)
        titulo.to_edge(UP)
        self.play(Write(titulo))
        self.wait(1)
        
        # Crear formas
        circulo = Circle(color=BLUE, radius=1)
        cuadrado = Square(color=RED, side_length=2)
        triangulo = Triangle(color=GREEN)
        
        # Agrupar y organizar
        formas = VGroup(circulo, cuadrado, triangulo)
        formas.arrange(RIGHT, buff=1)
        
        # Animar aparición
        self.play(Create(formas))
        self.wait(1)
        
        # Cambiar colores
        self.play(
            circulo.animate.set_fill(BLUE, opacity=0.5),
            cuadrado.animate.set_fill(RED, opacity=0.5),
            triangulo.animate.set_fill(GREEN, opacity=0.5)
        )
        self.wait(1)
        
        # Rotar
        self.play(Rotate(formas, angle=PI))
        self.wait(1)
        
        # Desaparecer
        self.play(FadeOut(formas), FadeOut(titulo))
        self.wait(1)
```
