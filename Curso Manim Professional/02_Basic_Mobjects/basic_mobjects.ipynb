## Dimensiones de la cámara

La cámara de Manim tiene dimensiones específicas que debes conocer:
- Por defecto: 14.22 × 8 unidades
- Centro: (0, 0, 0)
- Límites X: [-7.11, 7.11]
- Límites Y: [-4, 4]

```python
%%manim -qm -v Warning DimensionesCamara

class DimensionesCamara(Scene):
    def construct(self):
        # Mostrar los límites de la pantalla
        # Marco de referencia
        marco = Rectangle(width=14.22, height=8, color=WHITE)
        self.add(marco)
        
        # Marcar el centro
        centro = Dot(ORIGIN, color=RED, radius=0.1)
        centro_label = Text("(0, 0)", font_size=24).next_to(centro, DOWN)
        self.add(centro, centro_label)
        
        # Esquinas
        esquinas = [
            (UL, "(-7.11, 4)"),
            (UR, "(7.11, 4)"),
            (DL, "(-7.11, -4)"),
            (DR, "(7.11, -4)")
        ]
        
        for pos, label_text in esquinas:
            dot = Dot(pos, color=YELLOW, radius=0.08)
            label = Text(label_text, font_size=20).next_to(dot, pos/2)
            self.add(dot, label)
        
        self.wait(2)
```

```python
%%manim -qm -v Warning PosicionAbsoluta

class PosicionAbsoluta(Scene):
    def construct(self):
        # Vectores básicos de dirección
        # UP, DOWN, LEFT, RIGHT
        circulo_up = Circle(radius=0.3, color=BLUE).shift(UP * 2)
        circulo_down = Circle(radius=0.3, color=RED).shift(DOWN * 2)
        circulo_left = Circle(radius=0.3, color=GREEN).shift(LEFT * 3)
        circulo_right = Circle(radius=0.3, color=YELLOW).shift(RIGHT * 3)
        
        # Etiquetas
        label_up = Text("UP", font_size=20).next_to(circulo_up, UP)
        label_down = Text("DOWN", font_size=20).next_to(circulo_down, DOWN)
        label_left = Text("LEFT", font_size=20).next_to(circulo_left, LEFT)
        label_right = Text("RIGHT", font_size=20).next_to(circulo_right, RIGHT)
        
        # Centro
        centro = Circle(radius=0.3, color=WHITE)
        label_centro = Text("ORIGIN", font_size=20).next_to(centro, DOWN)
        
        self.play(
            Create(VGroup(circulo_up, circulo_down, circulo_left, circulo_right, centro)),
            Write(VGroup(label_up, label_down, label_left, label_right, label_centro))
        )
        self.wait(2)
```

### Combinación de vectores (Esquinas)

```python
%%manim -qm -v Warning Esquinas

class Esquinas(Scene):
    def construct(self):
        # UL = UP + LEFT, UR = UP + RIGHT
        # DL = DOWN + LEFT, DR = DOWN + RIGHT
        
        cuadrado_ul = Square(side_length=0.5, color=BLUE).move_to(UL * 2)
        cuadrado_ur = Square(side_length=0.5, color=RED).move_to(UR * 2)
        cuadrado_dl = Square(side_length=0.5, color=GREEN).move_to(DL * 2)
        cuadrado_dr = Square(side_length=0.5, color=YELLOW).move_to(DR * 2)
        
        labels = VGroup(
            Text("UL", font_size=24).next_to(cuadrado_ul, UL, buff=0.1),
            Text("UR", font_size=24).next_to(cuadrado_ur, UR, buff=0.1),
            Text("DL", font_size=24).next_to(cuadrado_dl, DL, buff=0.1),
            Text("DR", font_size=24).next_to(cuadrado_dr, DR, buff=0.1)
        )
        
        cuadrados = VGroup(cuadrado_ul, cuadrado_ur, cuadrado_dl, cuadrado_dr)
        
        self.play(Create(cuadrados))
        self.play(Write(labels))
        self.wait(2)
```

### Posición Relativa - to_edge()

```python
%%manim -qm -v Warning UsoToEdge

class UsoToEdge(Scene):
    def construct(self):
        # to_edge() coloca el objeto al borde de la pantalla
        titulo = Text("to_edge()", font_size=36, color=BLUE)
        titulo.to_edge(UP)
        
        texto_left = Text("LEFT", font_size=30, color=GREEN).to_edge(LEFT)
        texto_right = Text("RIGHT", font_size=30, color=RED).to_edge(RIGHT)
        texto_down = Text("DOWN", font_size=30, color=YELLOW).to_edge(DOWN)
        
        # Puedes agregar buff (buffer) para separación
        circulo_ul = Circle(radius=0.4, color=PURPLE).to_edge(UL, buff=0.5)
        
        self.play(Write(titulo))
        self.play(Write(VGroup(texto_left, texto_right, texto_down)))
        self.play(Create(circulo_ul))
        self.wait(2)
```

### Posición Relativa - next_to()

El método `next_to()` permite posicionar un objeto **junto a otro objeto**:
- Primer parámetro: el objeto de referencia
- Segundo parámetro: dirección (UP, DOWN, LEFT, RIGHT, etc.)
- `buff`: espacio entre objetos (por defecto 0.25)

```python
%%manim -qm -v Warning UsoNextTo

class UsoNextTo(Scene):
    def construct(self):
        # Objeto de referencia
        cuadrado = Square(side_length=2, color=BLUE)
        
        # Objetos posicionados con next_to()
        circulo_arriba = Circle(radius=0.5, color=RED).next_to(cuadrado, UP)
        circulo_abajo = Circle(radius=0.5, color=GREEN).next_to(cuadrado, DOWN)
        circulo_izq = Circle(radius=0.5, color=YELLOW).next_to(cuadrado, LEFT)
        circulo_der = Circle(radius=0.5, color=PURPLE).next_to(cuadrado, RIGHT)
        
        # Con buff personalizado
        triangulo = Triangle(color=ORANGE).scale(0.6)
        triangulo.next_to(cuadrado, UR, buff=0.8)
        
        self.play(Create(cuadrado))
        self.play(Create(VGroup(circulo_arriba, circulo_abajo, circulo_izq, circulo_der)))
        self.play(Create(triangulo))
        self.wait(2)
```

### Movimiento con shift()

El método `shift()` mueve un objeto **relativamente** desde su posición actual:
- Acepta vectores: `UP`, `DOWN`, `LEFT`, `RIGHT`, `ORIGIN`
- Se pueden combinar vectores: `UP + RIGHT`
- Puedes multiplicar por escalares: `LEFT * 2`

```python
%%manim -qm -v Warning UsoShift

class UsoShift(Scene):
    def construct(self):
        # Crear círculo en el centro
        circulo = Circle(radius=0.5, color=BLUE)
        self.play(Create(circulo))
        self.wait(0.5)
        
        # Mover hacia arriba
        self.play(circulo.animate.shift(UP * 2))
        self.wait(0.5)
        
        # Mover hacia la derecha
        self.play(circulo.animate.shift(RIGHT * 3))
        self.wait(0.5)
        
        # Mover en diagonal (combinando vectores)
        self.play(circulo.animate.shift(DOWN * 2 + LEFT * 2))
        self.wait(0.5)
        
        # Volver al centro
        self.play(circulo.animate.shift(LEFT))
        self.wait(2)
```

## Tamaño de los Mobjects

### Ancho y Alto (width y height)

Puedes modificar las dimensiones de los objetos:
- `.scale()`: escala proporcional (multiplica por factor)
- `.set_width()`: establece ancho específico
- `.set_height()`: establece alto específico
- `.stretch()`: estira en una dirección

```python
%%manim -qm -v Warning ManipulacionTamaño

class ManipulacionTamaño(Scene):
    def construct(self):
        # Crear varios cuadrados
        cuadrado1 = Square(color=BLUE)
        cuadrado2 = Square(color=GREEN)
        cuadrado3 = Square(color=RED)
        cuadrado4 = Square(color=YELLOW)
        
        # Posicionar inicialmente
        cuadrados = VGroup(cuadrado1, cuadrado2, cuadrado3, cuadrado4)
        cuadrados.arrange(RIGHT, buff=0.8)
        
        # Labels
        label1 = Text("Original", font_size=20).next_to(cuadrado1, DOWN)
        label2 = Text("scale(0.5)", font_size=20).next_to(cuadrado2, DOWN)
        label3 = Text("width=3", font_size=20).next_to(cuadrado3, DOWN)
        label4 = Text("height=3", font_size=20).next_to(cuadrado4, DOWN)
        
        self.play(Create(cuadrados))
        self.play(Write(VGroup(label1, label2, label3, label4)))
        self.wait(1)
        
        # Aplicar transformaciones
        self.play(cuadrado2.animate.scale(0.5))
        self.play(cuadrado3.animate.set_width(3))
        self.play(cuadrado4.animate.set_height(3))
        self.wait(2)
```

## Atributos de VMobjects

Los **VMobjects** (Vectorial Mobjects) tienen atributos especiales para color, trazo y relleno:

### Color y Trazo
- `.set_color()`: establece color general
- `.set_stroke()`: color y grosor del borde (stroke_color, stroke_width)
- `.set_fill()`: color y opacidad del relleno (fill_color, fill_opacity)
- `.set_opacity()`: opacidad general del objeto

```python
%%manim -qm -v Warning AtributosVMobject

class AtributosVMobject(Scene):
    def construct(self):
        # Círculos con diferentes estilos
        circulo1 = Circle(radius=0.8)
        circulo1.set_stroke(color=BLUE, width=5)
        circulo1.set_fill(color=RED, opacity=0.5)
        circulo1.shift(LEFT * 3)
        
        circulo2 = Circle(radius=0.8)
        circulo2.set_color(GREEN)
        circulo2.set_stroke(width=10)
        
        circulo3 = Circle(radius=0.8)
        circulo3.set_fill(color=YELLOW, opacity=0.9)
        circulo3.set_stroke(color=PURPLE, width=2)
        circulo3.shift(RIGHT * 3)
        
        # Labels
        label1 = Text("Stroke + Fill", font_size=20).next_to(circulo1, DOWN)
        label2 = Text("Color grueso", font_size=20).next_to(circulo2, DOWN)
        label3 = Text("Fill opaco", font_size=20).next_to(circulo3, DOWN)
        
        self.play(Create(VGroup(circulo1, circulo2, circulo3)))
        self.play(Write(VGroup(label1, label2, label3)))
        self.wait(2)
```

## Copias y Setters

### Método .copy()
Crea una **copia independiente** del objeto. Útil para:
- Crear duplicados sin afectar el original
- Transformaciones que requieren estado inicial y final

### Métodos Setter
Permiten modificar propiedades de forma encadenada:
- `.set_color(color)`: cambiar color
- `.set_opacity(value)`: cambiar opacidad (0-1)
- `.set_stroke(color, width)`: modificar borde
- `.set_fill(color, opacity)`: modificar relleno

```python
%%manim -qm -v Warning UsoCopias

class UsoCopias(Scene):
    def construct(self):
        # Crear cuadrado original
        cuadrado_original = Square(color=BLUE).shift(LEFT * 3)
        
        # Crear copia y modificarla
        cuadrado_copia = cuadrado_original.copy()
        cuadrado_copia.set_color(RED)
        cuadrado_copia.set_fill(YELLOW, opacity=0.7)
        cuadrado_copia.shift(RIGHT * 6)  # Mover desde la posición del original
        
        # Labels
        label_original = Text("Original", font_size=20).next_to(cuadrado_original, DOWN)
        label_copia = Text("Copia modificada", font_size=20).next_to(cuadrado_copia, DOWN)
        
        # Animación
        self.play(Create(cuadrado_original), Write(label_original))
        self.wait(1)
        
        # La copia aparece ya modificada
        self.play(Create(cuadrado_copia), Write(label_copia))
        self.wait(2)
        
        # Demostrar encadenamiento de setters
        circulo = Circle().set_color(PURPLE).set_fill(GREEN, 0.5).set_stroke(width=5)
        circulo.shift(DOWN * 2)
        self.play(Create(circulo))
        self.wait(2)
```
