````python
## Archivo de Configuración: manim.cfg

El archivo `manim.cfg` permite establecer configuraciones globales para tu proyecto. Se coloca en la raíz del proyecto.

### Estructura del manim.cfg

```ini
[CLI]
# Calidad de renderizado
quality = medium_quality

# Color de fondo
background_color = WHITE

# Directorio de salida
media_dir = ./media

# Archivos de salida
output_file = output

# Previsualización automática
preview = True

# Guardar último fotograma
save_last_frame = False

# Transparencia
transparent = False

# Frames por segundo
frame_rate = 60
```

### Niveles de Calidad
- `low_quality` o `-ql`: 480p, 15 fps
- `medium_quality` o `-qm`: 720p, 30 fps  
- `high_quality` o `-qh`: 1080p, 60 fps
- `production_quality` o `-qp`: 1440p, 60 fps
- `fourk_quality` o `-qk`: 2160p (4K), 60 fps
````

## Parámetros de Línea de Comandos (CLI)

Al renderizar con `manim` desde la terminal, puedes usar diversos flags:

### Comandos Básicos
```bash
# Renderizar con calidad media
manim -qm archivo.py NombreEscena

# Renderizar en alta calidad y mostrar preview
manim -qh -p archivo.py NombreEscena

# Renderizar con fondo transparente
manim -qm --transparent archivo.py NombreEscena

# Cambiar color de fondo
manim -qm -c BLACK archivo.py NombreEscena

# Renderizar solo último frame (imagen)
manim -qm -s archivo.py NombreEscena

# Guardar como GIF
manim -qm -i archivo.py NombreEscena
```

### Flags Importantes
- `-p` o `--preview`: Abre el video automáticamente
- `-s` o `--save_last_frame`: Guarda solo imagen final
- `-i` o `--save_as_gif`: Exporta como GIF
- `-c COLOR` o `--background_color COLOR`: Establece color de fondo
- `--transparent`: Fondo transparente
- `-r WxH` o `--resolution WxH`: Resolución personalizada

## Configuración con Diccionario config

Dentro del código Python, puedes modificar la configuración usando el diccionario `config`:

### Opciones Disponibles
```python
config.background_color = WHITE
config.frame_rate = 60
config.pixel_height = 1080
config.pixel_width = 1920
config.aspect_ratio = 16/9
```

### Ventajas
✅ Control programático de la configuración  
✅ Diferentes configuraciones por escena  
✅ Configuración dinámica basada en condiciones

````python
%%manim -qm -v Warning ConfiguracionBasica

class ConfiguracionBasica(Scene):
    def construct(self):
        # Esta escena usa la configuración por defecto desde %%manim
        # En este caso: -qm (medium quality)
        
        titulo = Text("Configuración Media", font_size=48, color=BLUE)
        subtitulo = Text("720p, 30 fps", font_size=30, color=GRAY)
        subtitulo.next_to(titulo, DOWN)
        
        info = VGroup(
            Text(f"Resolución: {config.pixel_width}x{config.pixel_height}", font_size=24),
            Text(f"FPS: {config.frame_rate}", font_size=24),
            Text(f"Fondo: {config.background_color}", font_size=24)
        )
        info.arrange(DOWN, aligned_edge=LEFT, buff=0.3)
        info.next_to(subtitulo, DOWN, buff=1)
        
        self.play(Write(titulo))
        self.play(FadeIn(subtitulo))
        self.play(Write(info))
        self.wait(2)
````

## Ejemplo: Color de Fondo Personalizado

Puedes cambiar el color de fondo usando el flag `-c` en el comando mágico de Jupyter:

````python
%%manim -qm -v Warning -c WHITE FondoBlanco

class FondoBlanco(Scene):
    def construct(self):
        # Escena con fondo blanco (definido en %%manim -c WHITE)
        titulo = Text("Fondo Blanco", font_size=48, color=BLACK)
        
        circulo = Circle(radius=1.5, color=BLUE)
        circulo.set_stroke(width=5)
        circulo.set_fill(RED, opacity=0.5)
        circulo.next_to(titulo, DOWN, buff=1)
        
        self.play(Write(titulo))
        self.play(Create(circulo))
        self.wait(2)
````

## Modificando config Dentro del Código

Puedes modificar `config` directamente en tu clase Scene para cambios específicos:

````python
%%manim -qm -v Warning ConfigEnCodigo

class ConfigEnCodigo(Scene):
    def construct(self):
        # Modificar config dentro del código
        # Nota: algunos cambios deben hacerse antes de crear objetos
        
        # Mostrar configuración actual
        titulo = Text("Configuración Programática", font_size=40, color=YELLOW)
        titulo.to_edge(UP)
        
        info_config = VGroup(
            Text(f"Frame Width: {config.frame_width}", font_size=24),
            Text(f"Frame Height: {config.frame_height}", font_size=24),
            Text(f"Pixel Width: {config.pixel_width}", font_size=24),
            Text(f"Pixel Height: {config.pixel_height}", font_size=24),
            Text(f"Frame Rate: {config.frame_rate} fps", font_size=24),
            Text(f"Background: {config.background_color}", font_size=24)
        )
        info_config.arrange(DOWN, aligned_edge=LEFT, buff=0.25)
        info_config.next_to(titulo, DOWN, buff=0.8)
        
        self.play(Write(titulo))
        self.play(Write(info_config), run_time=2)
        self.wait(2)
````

## Ejemplo: Fondo Transparente

Para crear videos con fondo transparente (útil para overlays), usa el flag `--transparent`:

````python
%%manim -qm -v Warning --transparent FondoTransparente

class FondoTransparente(Scene):
    def construct(self):
        # Esta escena se renderiza con fondo transparente
        # Ideal para composiciones en editores de video
        
        titulo = Text("Fondo Transparente", font_size=44, color=BLUE)
        titulo.set_stroke(WHITE, width=2, background=True)
        
        formas = VGroup(
            Circle(radius=0.8, color=RED).set_fill(RED, 0.7),
            Square(side_length=1.5, color=GREEN).set_fill(GREEN, 0.7),
            Triangle(color=YELLOW).scale(1.2).set_fill(YELLOW, 0.7)
        )
        formas.arrange(RIGHT, buff=1)
        formas.next_to(titulo, DOWN, buff=1)
        
        self.play(Write(titulo))
        self.play(Create(formas), run_time=2)
        self.wait(2)
````

## Resolución y Aspect Ratio Personalizados

Puedes definir resoluciones personalizadas modificando `config` o usando flags CLI:

````python
%%manim -qm -v Warning ResolucionPersonalizada

class ResolucionPersonalizada(Scene):
    def construct(self):
        # Puedes cambiar la resolución antes de crear la escena
        # usando: config.pixel_width = 1920
        #         config.pixel_height = 1080
        
        titulo = Text("Resolución Actual", font_size=40, color=GREEN)
        titulo.to_edge(UP)
        
        # Información de resolución
        resolucion_info = VGroup(
            Text(f"Ancho: {config.pixel_width} px", font_size=30),
            Text(f"Alto: {config.pixel_height} px", font_size=30),
            Text(f"Aspect Ratio: {config.aspect_ratio:.2f}", font_size=30)
        )
        resolucion_info.arrange(DOWN, buff=0.4)
        resolucion_info.next_to(titulo, DOWN, buff=1)
        
        # Rectángulo representando la pantalla
        pantalla = Rectangle(
            width=config.frame_width * 0.6,
            height=config.frame_height * 0.6,
            color=BLUE
        )
        pantalla.next_to(resolucion_info, DOWN, buff=0.8)
        
        self.play(Write(titulo))
        self.play(Write(resolucion_info))
        self.play(Create(pantalla))
        self.wait(2)
````

## Frame Rate (FPS)

El frame rate determina la fluidez de la animación:

### Configuración de FPS
```python
config.frame_rate = 60  # 60 fps (más fluido)
config.frame_rate = 30  # 30 fps (estándar)
config.frame_rate = 24  # 24 fps (cinematográfico)
```

**Importante:** Mayor FPS = mayor tiempo de renderizado y mayor tamaño de archivo.

````python
%%manim -qm -v Warning DemoFPS

class DemoFPS(Scene):
    def construct(self):
        titulo = Text(f"Frame Rate: {config.frame_rate} FPS", font_size=40, color=ORANGE)
        titulo.to_edge(UP)
        
        # Círculo que rota constantemente
        circulo = Circle(radius=2, color=BLUE)
        circulo.set_stroke(width=5)
        
        # Punto en el perímetro para visualizar rotación
        punto = Dot(color=RED, radius=0.15)
        punto.move_to(circulo.point_from_proportion(0))
        
        info = Text(
            f"A {config.frame_rate} fps, la animación se ve {'muy fluida' if config.frame_rate >= 60 else 'fluida' if config.frame_rate >= 30 else 'normal'}",
            font_size=24
        )
        info.to_edge(DOWN)
        
        self.play(Write(titulo))
        self.play(Create(circulo), Create(punto))
        self.play(Write(info))
        
        # Rotación para mostrar fluidez
        self.play(
            Rotate(circulo, angle=2*PI, about_point=ORIGIN),
            MoveAlongPath(punto, circulo),
            run_time=3,
            rate_func=linear
        )
        self.wait(1)
````

## Opciones Avanzadas de config

### Propiedades Útiles del Diccionario config

```python
# Dimensiones de frame (espacio de coordenadas)
config.frame_width    # Ancho en unidades Manim (default: 14.22)
config.frame_height   # Alto en unidades Manim (default: 8.0)
config.frame_x_radius # Radio X desde el centro
config.frame_y_radius # Radio Y desde el centro

# Dimensiones de píxeles (resolución de salida)
config.pixel_width    # Ancho en píxeles
config.pixel_height   # Alto en píxeles

# Temporización
config.frame_rate     # Frames por segundo

# Apariencia
config.background_color      # Color de fondo
config.background_opacity    # Opacidad del fondo (para transparent)

# Directorios
config.media_dir        # Directorio de salida
config.video_dir        # Carpeta de videos
config.images_dir       # Carpeta de imágenes
config.text_dir         # Carpeta de archivos LaTeX

# Renderizado
config.quality          # Calidad (low, medium, high, etc.)
config.preview          # Abrir preview automáticamente
config.write_to_movie   # Escribir archivo de video
config.save_last_frame  # Guardar último frame como imagen
```

````python
%%manim -qm -v Warning ConfigCompleto

class ConfigCompleto(Scene):
    def construct(self):
        titulo = Text("Información Completa de Config", font_size=36, color=TEAL)
        titulo.to_edge(UP)
        
        # Información dividida en columnas
        col1 = VGroup(
            Text("DIMENSIONES", font_size=22, color=YELLOW),
            Text(f"Frame: {config.frame_width:.2f} × {config.frame_height:.2f}", font_size=18),
            Text(f"Píxeles: {config.pixel_width} × {config.pixel_height}", font_size=18),
            Text(f"Aspect: {config.aspect_ratio:.3f}", font_size=18),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.2)
        
        col2 = VGroup(
            Text("RENDERIZADO", font_size=22, color=YELLOW),
            Text(f"FPS: {config.frame_rate}", font_size=18),
            Text(f"Calidad: {config.quality}", font_size=18),
            Text(f"Fondo: {config.background_color}", font_size=18),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.2)
        
        col3 = VGroup(
            Text("DIRECTORIOS", font_size=22, color=YELLOW),
            Text(f"Media: .../{config.media_dir.split('/')[-1]}", font_size=18),
            Text(f"Preview: {config.preview}", font_size=18),
            Text(f"Save frame: {config.save_last_frame}", font_size=18),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.2)
        
        columnas = VGroup(col1, col2, col3).arrange(RIGHT, buff=1, aligned_edge=UP)
        columnas.next_to(titulo, DOWN, buff=0.7)
        
        self.play(Write(titulo))
        self.play(Write(col1))
        self.play(Write(col2))
        self.play(Write(col3))
        self.wait(3)
````
