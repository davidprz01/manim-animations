## Concepto de Capas (Layers)

En Manim, cuando se superponen múltiples objetos, el **orden de renderizado** determina cuál aparece delante de otro.

### Regla por Defecto
- Los objetos se renderizan en el **orden en que fueron agregados** a la escena
- El último objeto agregado aparece **delante** de los anteriores
- Esto es similar al concepto de "capas" en software de diseño gráfico

### Atributo z_index
- `z_index = 0` es el valor por defecto
- **Valores mayores** = más adelante (más cerca del espectador)
- **Valores menores** = más atrás (más lejos del espectador)

````python
%%manim -qm -v Warning OrdenPorDefecto

class OrdenPorDefecto(Scene):
    def construct(self):
        titulo = Text("Orden por Defecto", font_size=40, color=WHITE)
        titulo.to_edge(UP)
        
        # Crear tres círculos superpuestos
        # El orden de add() determina quién aparece adelante
        circulo_rojo = Circle(radius=1.5, color=RED).set_fill(RED, opacity=0.8)
        circulo_verde = Circle(radius=1.5, color=GREEN).set_fill(GREEN, opacity=0.8)
        circulo_azul = Circle(radius=1.5, color=BLUE).set_fill(BLUE, opacity=0.8)
        
        circulo_rojo.shift(LEFT * 1.5)
        circulo_verde.move_to(ORIGIN)
        circulo_azul.shift(RIGHT * 1.5)
        
        # Labels
        label1 = Text("1º agregado", font_size=20).next_to(circulo_rojo, DOWN, buff=1.5)
        label2 = Text("2º agregado", font_size=20).next_to(circulo_verde, DOWN, buff=1.5)
        label3 = Text("3º agregado (adelante)", font_size=20).next_to(circulo_azul, DOWN, buff=1.5)
        
        self.play(Write(titulo))
        
        # El orden de animación determina el z-order
        self.play(Create(circulo_rojo))
        self.play(Create(circulo_verde))
        self.play(Create(circulo_azul))
        
        self.play(Write(VGroup(label1, label2, label3)))
        self.wait(2)
````

## Usando z_index

El atributo `z_index` permite controlar explícitamente el orden de renderizado sin depender del orden de agregado.

```python
objeto.z_index = 1      # Adelante
objeto.z_index = 0      # Normal (default)
objeto.z_index = -1     # Atrás
```

También puedes usar el método `.set_z_index(valor)`:

````python
%%manim -qm -v Warning UsoZIndex

class UsoZIndex(Scene):
    def construct(self):
        titulo = Text("Control con z_index", font_size=40, color=WHITE)
        titulo.to_edge(UP)
        
        # Crear tres cuadrados superpuestos
        cuadrado_rojo = Square(side_length=2.5, color=RED).set_fill(RED, opacity=0.9)
        cuadrado_verde = Square(side_length=2.5, color=GREEN).set_fill(GREEN, opacity=0.9)
        cuadrado_azul = Square(side_length=2.5, color=BLUE).set_fill(BLUE, opacity=0.9)
        
        # Posicionar superpuestos
        cuadrado_rojo.shift(LEFT * 1 + UP * 0.5)
        cuadrado_verde.move_to(ORIGIN)
        cuadrado_azul.shift(RIGHT * 1 + DOWN * 0.5)
        
        # Asignar z_index: el rojo adelante, el azul atrás
        cuadrado_rojo.z_index = 2   # Más adelante
        cuadrado_verde.z_index = 1  # Medio
        cuadrado_azul.z_index = 0   # Más atrás
        
        # Labels con z_index info
        label_rojo = Text("z_index = 2", font_size=18, color=WHITE)
        label_rojo.move_to(cuadrado_rojo)
        label_verde = Text("z_index = 1", font_size=18, color=WHITE)
        label_verde.move_to(cuadrado_verde)
        label_azul = Text("z_index = 0", font_size=18, color=WHITE)
        label_azul.move_to(cuadrado_azul)
        
        # Labels heredan z_index de sus padres
        label_rojo.z_index = 3
        label_verde.z_index = 3
        label_azul.z_index = 3
        
        self.play(Write(titulo))
        # Agregamos en orden inverso para demostrar que z_index domina
        self.play(Create(cuadrado_azul))  # Primero azul
        self.play(Create(cuadrado_verde)) # Luego verde
        self.play(Create(cuadrado_rojo))  # Luego rojo (pero z_index=2 lo pone adelante)
        
        self.play(Write(VGroup(label_rojo, label_verde, label_azul)))
        self.wait(2)
````

## Métodos bring_to_front() y bring_to_back()

Estos métodos de la Scene permiten cambiar el orden de los objetos dinámicamente:

- `self.bring_to_front(objeto)`: Trae el objeto al frente de todo
- `self.bring_to_back(objeto)`: Envía el objeto atrás de todo

**Nota:** Estos métodos modifican el orden de renderizado, no el `z_index`.

````python
%%manim -qm -v Warning BringToFrontBack

class BringToFrontBack(Scene):
    def construct(self):
        titulo = Text("bring_to_front() y bring_to_back()", font_size=36)
        titulo.to_edge(UP)
        
        # Crear tres círculos superpuestos
        circulo_rojo = Circle(radius=1.2, color=RED).set_fill(RED, opacity=0.9)
        circulo_verde = Circle(radius=1.2, color=GREEN).set_fill(GREEN, opacity=0.9)
        circulo_azul = Circle(radius=1.2, color=BLUE).set_fill(BLUE, opacity=0.9)
        
        circulo_rojo.shift(LEFT)
        circulo_azul.shift(RIGHT)
        
        # Labels
        texto_info = Text("Rojo está adelante", font_size=24).to_edge(DOWN)
        
        self.play(Write(titulo))
        self.play(Create(circulo_rojo), Create(circulo_verde), Create(circulo_azul))
        self.play(Write(texto_info))
        self.wait(1)
        
        # Traer verde al frente
        self.bring_to_front(circulo_verde)
        self.play(
            Transform(texto_info, Text("Verde traído al frente", font_size=24).to_edge(DOWN))
        )
        self.wait(1)
        
        # Enviar verde atrás
        self.bring_to_back(circulo_verde)
        self.play(
            Transform(texto_info, Text("Verde enviado atrás", font_size=24).to_edge(DOWN))
        )
        self.wait(1)
        
        # Traer azul al frente
        self.bring_to_front(circulo_azul)
        self.play(
            Transform(texto_info, Text("Azul traído al frente", font_size=24).to_edge(DOWN))
        )
        self.wait(2)
````

## Animando cambios de z_index

Puedes animar el cambio de `z_index` usando `.animate` para crear efectos visuales interesantes:

````python
%%manim -qm -v Warning AnimandoZIndex

class AnimandoZIndex(Scene):
    def construct(self):
        titulo = Text("Animando z_index", font_size=40)
        titulo.to_edge(UP)
        
        # Crear cartas superpuestas
        cartas = VGroup()
        colores = [RED, ORANGE, YELLOW, GREEN, BLUE, PURPLE]
        
        for i, color in enumerate(colores):
            carta = RoundedRectangle(
                corner_radius=0.1,
                width=1.5,
                height=2,
                color=color
            ).set_fill(color, opacity=0.9)
            carta.shift(RIGHT * i * 0.4)
            carta.z_index = i
            cartas.add(carta)
        
        cartas.center()
        
        self.play(Write(titulo))
        self.play(Create(cartas), run_time=2)
        self.wait(1)
        
        # Animar: traer primera carta al frente
        primera_carta = cartas[0]
        self.play(primera_carta.animate.shift(RIGHT * 3))
        self.bring_to_front(primera_carta)
        self.play(primera_carta.animate.shift(LEFT * 3))
        self.wait(1)
        
        # Reorganizar todas las cartas
        for i, carta in enumerate(reversed(cartas)):
            self.bring_to_front(carta)
            self.wait(0.2)
        
        self.wait(2)
````

## Ejemplo Práctico: Diagrama de Venn

Un caso de uso común de z_index es en diagramas de Venn donde necesitas controlar qué círculos se superponen y mostrar intersecciones:

````python
%%manim -qm -v Warning DiagramaVenn

class DiagramaVenn(Scene):
    def construct(self):
        titulo = Text("Diagrama de Venn", font_size=40)
        titulo.to_edge(UP)
        
        # Crear tres círculos con opacidad para ver intersecciones
        circulo_a = Circle(radius=1.5, color=RED)
        circulo_a.set_fill(RED, opacity=0.4)
        circulo_a.set_stroke(RED, width=3)
        circulo_a.shift(LEFT * 1 + UP * 0.5)
        
        circulo_b = Circle(radius=1.5, color=GREEN)
        circulo_b.set_fill(GREEN, opacity=0.4)
        circulo_b.set_stroke(GREEN, width=3)
        circulo_b.shift(RIGHT * 1 + UP * 0.5)
        
        circulo_c = Circle(radius=1.5, color=BLUE)
        circulo_c.set_fill(BLUE, opacity=0.4)
        circulo_c.set_stroke(BLUE, width=3)
        circulo_c.shift(DOWN * 0.8)
        
        # Labels para cada conjunto
        label_a = Text("A", font_size=30, color=RED).move_to(circulo_a.get_center() + UL * 0.8)
        label_b = Text("B", font_size=30, color=GREEN).move_to(circulo_b.get_center() + UR * 0.8)
        label_c = Text("C", font_size=30, color=BLUE).move_to(circulo_c.get_center() + DOWN * 0.8)
        
        # Los labels deben estar adelante
        label_a.z_index = 10
        label_b.z_index = 10
        label_c.z_index = 10
        
        grupo_venn = VGroup(circulo_a, circulo_b, circulo_c)
        grupo_venn.shift(DOWN * 0.3)
        
        self.play(Write(titulo))
        self.play(Create(circulo_a), Create(circulo_b), Create(circulo_c))
        self.play(Write(VGroup(label_a, label_b, label_c)))
        self.wait(2)
        
        # Destacar un círculo trayéndolo al frente
        self.play(circulo_a.animate.set_fill(opacity=0.8))
        self.bring_to_front(circulo_a)
        self.wait(1)
        
        # Restaurar
        self.play(circulo_a.animate.set_fill(opacity=0.4))
        self.wait(2)
````

## Z-Index en VGroups

Cuando trabajas con VGroups, el `z_index` se aplica a todo el grupo. Los elementos dentro del grupo se renderizan en su orden interno.

````python
%%manim -qm -v Warning ZIndexVGroup

class ZIndexVGroup(Scene):
    def construct(self):
        titulo = Text("Z-Index en VGroups", font_size=40)
        titulo.to_edge(UP)
        
        # Grupo 1: Cuadrados (z_index bajo)
        grupo_cuadrados = VGroup(
            Square(side_length=1.5, color=RED).set_fill(RED, 0.8),
            Square(side_length=1.5, color=ORANGE).set_fill(ORANGE, 0.8),
            Square(side_length=1.5, color=YELLOW).set_fill(YELLOW, 0.8)
        )
        grupo_cuadrados.arrange(RIGHT, buff=0.5)
        grupo_cuadrados.shift(UP * 0.5)
        grupo_cuadrados.z_index = 0  # Atrás
        
        # Grupo 2: Círculos (z_index alto)
        grupo_circulos = VGroup(
            Circle(radius=0.8, color=BLUE).set_fill(BLUE, 0.8),
            Circle(radius=0.8, color=PURPLE).set_fill(PURPLE, 0.8),
            Circle(radius=0.8, color=PINK).set_fill(PINK, 0.8)
        )
        grupo_circulos.arrange(RIGHT, buff=0.5)
        grupo_circulos.shift(DOWN * 0.3)
        grupo_circulos.z_index = 1  # Adelante
        
        # Labels
        label_cuadrados = Text("Grupo z_index = 0 (atrás)", font_size=20)
        label_cuadrados.next_to(grupo_cuadrados, DOWN, buff=2)
        label_circulos = Text("Grupo z_index = 1 (adelante)", font_size=20)
        label_circulos.next_to(grupo_circulos, DOWN, buff=0.5)
        
        self.play(Write(titulo))
        self.play(Create(grupo_cuadrados), Create(grupo_circulos))
        self.play(Write(label_cuadrados), Write(label_circulos))
        self.wait(2)
        
        # Intercambiar z_index
        self.play(
            Transform(label_cuadrados, Text("Ahora z_index = 2 (adelante)", font_size=20).move_to(label_cuadrados)),
            Transform(label_circulos, Text("Ahora z_index = 0 (atrás)", font_size=20).move_to(label_circulos))
        )
        grupo_cuadrados.z_index = 2
        grupo_circulos.z_index = 0
        self.wait(2)
````

## Ejemplo Avanzado: Efecto de Profundidad

Combinando `z_index` con `scale` y `opacity` puedes crear efectos de profundidad visual:

````python
%%manim -qm -v Warning EfectoProfundidad

class EfectoProfundidad(Scene):
    def construct(self):
        titulo = Text("Efecto de Profundidad", font_size=40)
        titulo.to_edge(UP)
        titulo.z_index = 100  # Siempre visible
        
        # Crear varios cuadrados con diferentes "profundidades"
        capas = VGroup()
        
        for i in range(5):
            cuadrado = Square(
                side_length=3 - i * 0.4,
                color=interpolate_color(BLUE, WHITE, i / 4)
            )
            cuadrado.set_fill(interpolate_color(BLUE, BLUE_E, i / 4), opacity=0.7 - i * 0.1)
            cuadrado.z_index = 4 - i  # Más atrás = menor z_index
            cuadrado.shift(DOWN * 0.3 + LEFT * 0.2 * i)
            capas.add(cuadrado)
        
        capas.center()
        
        self.play(Write(titulo))
        
        # Animar aparición de capas desde atrás hacia adelante
        for i, capa in enumerate(reversed(capas)):
            self.play(Create(capa), run_time=0.4)
        
        self.wait(1)
        
        # Efecto de "hover" sobre la primera capa
        capa_frontal = capas[0]
        self.play(
            capa_frontal.animate.scale(1.1).set_fill(opacity=0.9),
            run_time=0.5
        )
        self.play(
            capa_frontal.animate.scale(1/1.1).set_fill(opacity=0.7),
            run_time=0.5
        )
        self.wait(2)
````

✅ **set_z_index()**: Método para asignar z_index  
✅ **bring_to_front()**: Trae un objeto al frente de la escena  
✅ **bring_to_back()**: Envía un objeto detrás de todos  
✅ **VGroups y z_index**: El z_index del grupo afecta a todos sus elementos  
✅ **Efectos de profundidad**: Combinar z_index con opacity y scale
