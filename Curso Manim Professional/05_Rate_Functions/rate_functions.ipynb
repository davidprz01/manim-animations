## ¿Qué es una Rate Function?

Una **rate function** es una función matemática que mapea el tiempo de animación (0 a 1) a un progreso de animación (también 0 a 1).

### Ejemplo Conceptual
```
t=0.0 → Inicio de la animación
t=0.5 → Mitad del tiempo
t=1.0 → Fin de la animación
```

Sin embargo, con rate functions, el **progreso visual** puede ser diferente al tiempo transcurrido:
- `smooth`: Comienza lento, acelera en el medio, desacelera al final
- `linear`: Velocidad constante
- `there_and_back`: Va al destino y regresa

### Sintaxis
```python
self.play(animacion, rate_func=nombre_rate_function)
```

## Rate Functions Básicas

### smooth (default)
La función por defecto en Manim. Proporciona una transición suave con aceleración y desaceleración.

````python
%%manim -qm -v Warning ComparacionBasica

class ComparacionBasica(Scene):
    def construct(self):
        titulo = Text("Comparación: smooth vs linear", font_size=36)
        titulo.to_edge(UP)
        
        # Crear dos círculos para comparar
        circulo_smooth = Circle(radius=0.5, color=BLUE).set_fill(BLUE, 0.8)
        circulo_linear = Circle(radius=0.5, color=RED).set_fill(RED, 0.8)
        
        circulo_smooth.shift(LEFT * 5 + UP)
        circulo_linear.shift(LEFT * 5 + DOWN)
        
        # Labels
        label_smooth = Text("smooth (default)", font_size=24).next_to(circulo_smooth, LEFT)
        label_linear = Text("linear", font_size=24).next_to(circulo_linear, LEFT)
        
        # Líneas guía
        linea_arriba = DashedLine(LEFT * 5 + UP, RIGHT * 5 + UP, color=GRAY)
        linea_abajo = DashedLine(LEFT * 5 + DOWN, RIGHT * 5 + DOWN, color=GRAY)
        
        self.play(Write(titulo))
        self.add(linea_arriba, linea_abajo)
        self.play(Create(circulo_smooth), Create(circulo_linear))
        self.play(Write(label_smooth), Write(label_linear))
        self.wait(1)
        
        # Animación comparativa
        self.play(
            circulo_smooth.animate.shift(RIGHT * 10),
            circulo_linear.animate.shift(RIGHT * 10),
            rate_func=smooth,  # Ambos usarán smooth por defecto
            run_time=3
        )
        self.wait(0.5)
        
        # Reiniciar posición
        circulo_smooth.shift(LEFT * 10)
        circulo_linear.shift(LEFT * 10)
        
        # Ahora con diferentes rate_func
        self.play(
            circulo_smooth.animate.shift(RIGHT * 10),
            run_time=3,
            rate_func=smooth
        )
        circulo_smooth.shift(LEFT * 10)
        
        self.play(
            circulo_linear.animate.shift(RIGHT * 10),
            run_time=3,
            rate_func=linear
        )
        self.wait(2)
````

## Catálogo de Rate Functions

Manim incluye muchas rate functions predefinidas. Veamos las más importantes:

````python
%%manim -qm -v Warning CatalogoRateFunctions

class CatalogoRateFunctions(Scene):
    def construct(self):
        titulo = Text("Catálogo de Rate Functions", font_size=36)
        titulo.to_edge(UP)
        self.play(Write(titulo))
        
        # Lista de rate functions a demostrar
        rate_functions = [
            ("linear", linear),
            ("smooth", smooth),
            ("rush_into", rush_into),
            ("rush_from", rush_from),
            ("slow_into", slow_into),
            ("double_smooth", double_smooth),
        ]
        
        # Crear elementos para cada rate function
        elementos = VGroup()
        for i, (nombre, _) in enumerate(rate_functions):
            dot = Dot(color=interpolate_color(BLUE, RED, i / len(rate_functions)))
            dot.shift(LEFT * 5)
            label = Text(nombre, font_size=20).next_to(dot, LEFT, buff=0.3)
            elementos.add(VGroup(dot, label))
        
        elementos.arrange(DOWN, buff=0.4, aligned_edge=LEFT)
        elementos.next_to(titulo, DOWN, buff=0.5)
        
        # Línea guía
        for elem in elementos:
            linea = DashedLine(LEFT * 5, RIGHT * 5, color=GRAY, stroke_opacity=0.3)
            linea.move_to(elem[0].get_center())
            self.add(linea)
        
        self.play(Create(elementos))
        self.wait(1)
        
        # Animar cada punto con su rate function
        animaciones = []
        for i, (nombre, rate_func) in enumerate(rate_functions):
            dot = elementos[i][0]
            animaciones.append(
                dot.animate.shift(RIGHT * 10)
            )
        
        # Ejecutar todas simultáneamente con diferentes rate_funcs
        self.play(
            elementos[0][0].animate.shift(RIGHT * 10),
            rate_func=linear, run_time=3
        )
        elementos[0][0].shift(LEFT * 10)
        
        self.play(
            elementos[1][0].animate.shift(RIGHT * 10),
            rate_func=smooth, run_time=3
        )
        elementos[1][0].shift(LEFT * 10)
        
        self.play(
            elementos[2][0].animate.shift(RIGHT * 10),
            rate_func=rush_into, run_time=3
        )
        
        self.wait(2)
````

## Ease In, Ease Out, Ease In-Out

Las rate functions de **easing** controlan la aceleración:

- **ease_in_***: Comienza lento, acelera (entra suave)
- **ease_out_***: Comienza rápido, desacelera (sale suave)
- **ease_in_out_***: Combina ambos (suave en ambos extremos)

````python
%%manim -qm -v Warning EasingFunctions

class EasingFunctions(Scene):
    def construct(self):
        titulo = Text("Funciones de Easing", font_size=36)
        titulo.to_edge(UP)
        
        # Crear tres cuadrados
        cuadrado_in = Square(side_length=0.8, color=RED).set_fill(RED, 0.8)
        cuadrado_out = Square(side_length=0.8, color=GREEN).set_fill(GREEN, 0.8)
        cuadrado_inout = Square(side_length=0.8, color=BLUE).set_fill(BLUE, 0.8)
        
        # Posicionar
        cuadrado_in.shift(LEFT * 5 + UP * 1.5)
        cuadrado_out.shift(LEFT * 5)
        cuadrado_inout.shift(LEFT * 5 + DOWN * 1.5)
        
        # Labels
        label_in = Text("ease_in_sine", font_size=20, color=RED)
        label_out = Text("ease_out_sine", font_size=20, color=GREEN)
        label_inout = Text("ease_in_out_sine", font_size=20, color=BLUE)
        
        label_in.next_to(cuadrado_in, DOWN, buff=0.2)
        label_out.next_to(cuadrado_out, DOWN, buff=0.2)
        label_inout.next_to(cuadrado_inout, DOWN, buff=0.2)
        
        self.play(Write(titulo))
        self.play(
            Create(VGroup(cuadrado_in, cuadrado_out, cuadrado_inout)),
            Write(VGroup(label_in, label_out, label_inout))
        )
        self.wait(1)
        
        # Animación simultánea con diferentes easing
        self.play(
            cuadrado_in.animate.shift(RIGHT * 10),
            cuadrado_out.animate.shift(RIGHT * 10),
            cuadrado_inout.animate.shift(RIGHT * 10),
            label_in.animate.shift(RIGHT * 10),
            label_out.animate.shift(RIGHT * 10),
            label_inout.animate.shift(RIGHT * 10),
            run_time=3
        )
        self.wait(1)
        
        # Reiniciar
        self.play(
            cuadrado_in.animate.shift(LEFT * 10),
            cuadrado_out.animate.shift(LEFT * 10),
            cuadrado_inout.animate.shift(LEFT * 10),
            label_in.animate.shift(LEFT * 10),
            label_out.animate.shift(LEFT * 10),
            label_inout.animate.shift(LEFT * 10),
            run_time=0.5
        )
        
        # Mostrar diferencias una por una
        info = Text("ease_in: comienza lento", font_size=24).to_edge(DOWN)
        self.play(Write(info))
        self.play(cuadrado_in.animate.shift(RIGHT * 10), rate_func=ease_in_sine, run_time=2)
        
        self.play(Transform(info, Text("ease_out: termina lento", font_size=24).to_edge(DOWN)))
        self.play(cuadrado_out.animate.shift(RIGHT * 10), rate_func=ease_out_sine, run_time=2)
        
        self.play(Transform(info, Text("ease_in_out: suave en ambos extremos", font_size=24).to_edge(DOWN)))
        self.play(cuadrado_inout.animate.shift(RIGHT * 10), rate_func=ease_in_out_sine, run_time=2)
        
        self.wait(2)
````

## Funciones Especiales: there_and_back

`there_and_back` es perfecta para efectos de "pulso" o "vibración" donde el objeto va al destino y regresa:

````python
%%manim -qm -v Warning ThereAndBack

class ThereAndBack(Scene):
    def construct(self):
        titulo = Text("there_and_back", font_size=40)
        titulo.to_edge(UP)
        
        # Círculo para demostración
        circulo = Circle(radius=1, color=BLUE).set_fill(BLUE, 0.7)
        
        # Texto explicativo
        explicacion = Text("El objeto va al destino y regresa", font_size=24)
        explicacion.to_edge(DOWN)
        
        self.play(Write(titulo))
        self.play(Create(circulo))
        self.play(Write(explicacion))
        self.wait(1)
        
        # Demostración: escalar con there_and_back
        self.play(
            circulo.animate.scale(2),
            rate_func=there_and_back,
            run_time=2
        )
        self.wait(0.5)
        
        # Demostración: mover con there_and_back
        self.play(
            circulo.animate.shift(RIGHT * 3),
            rate_func=there_and_back,
            run_time=2
        )
        self.wait(0.5)
        
        # Demostración: cambiar color con there_and_back
        self.play(
            circulo.animate.set_color(RED),
            rate_func=there_and_back,
            run_time=2
        )
        self.wait(0.5)
        
        # Variante: there_and_back_with_pause
        nuevo_texto = Text("there_and_back_with_pause: pausa en el destino", font_size=22)
        nuevo_texto.to_edge(DOWN)
        self.play(Transform(explicacion, nuevo_texto))
        
        self.play(
            circulo.animate.scale(1.5).shift(UP),
            rate_func=there_and_back_with_pause,
            run_time=3
        )
        self.wait(2)
````

## Efectos de Rebote: Bounce

Las funciones `bounce` simulan el efecto físico de un objeto rebotando:

````python
%%manim -qm -v Warning EfectosBounce

class EfectosBounce(Scene):
    def construct(self):
        titulo = Text("Efectos de Rebote", font_size=40)
        titulo.to_edge(UP)
        
        # Crear pelotas
        pelota1 = Circle(radius=0.6, color=RED).set_fill(RED, 0.9)
        pelota2 = Circle(radius=0.6, color=GREEN).set_fill(GREEN, 0.9)
        pelota3 = Circle(radius=0.6, color=BLUE).set_fill(BLUE, 0.9)
        
        pelota1.shift(LEFT * 4 + UP * 3)
        pelota2.shift(UP * 3)
        pelota3.shift(RIGHT * 4 + UP * 3)
        
        # Labels
        label1 = Text("ease_out_bounce", font_size=18, color=RED).shift(LEFT * 4 + DOWN * 2)
        label2 = Text("ease_in_bounce", font_size=18, color=GREEN).shift(DOWN * 2)
        label3 = Text("ease_in_out_bounce", font_size=18, color=BLUE).shift(RIGHT * 4 + DOWN * 2)
        
        # Suelo
        suelo = Line(LEFT * 6 + DOWN * 1.5, RIGHT * 6 + DOWN * 1.5, color=WHITE)
        
        self.play(Write(titulo))
        self.add(suelo)
        self.play(Create(VGroup(pelota1, pelota2, pelota3)))
        self.play(Write(VGroup(label1, label2, label3)))
        self.wait(1)
        
        # Animación de caída con bounce
        self.play(
            pelota1.animate.shift(DOWN * 4.5),
            pelota2.animate.shift(DOWN * 4.5),
            pelota3.animate.shift(DOWN * 4.5),
            run_time=2,
            rate_func=ease_out_bounce
        )
        self.wait(1)
        
        # Reiniciar y mostrar cada uno
        pelota1.shift(UP * 4.5)
        pelota2.shift(UP * 4.5)
        pelota3.shift(UP * 4.5)
        
        self.play(pelota1.animate.shift(DOWN * 4.5), rate_func=ease_out_bounce, run_time=2)
        self.play(pelota2.animate.shift(DOWN * 4.5), rate_func=ease_in_bounce, run_time=2)
        self.play(pelota3.animate.shift(DOWN * 4.5), rate_func=ease_in_out_bounce, run_time=2)
        
        self.wait(2)
````

## Efectos Elásticos: Elastic

Las funciones `elastic` crean un efecto de "resorte" que sobrepasa el destino y oscila:

````python
%%manim -qm -v Warning EfectosElastic

class EfectosElastic(Scene):
    def construct(self):
        titulo = Text("Efectos Elásticos", font_size=40)
        titulo.to_edge(UP)
        
        # Crear cuadrados
        cuadrado1 = Square(side_length=1.5, color=ORANGE).set_fill(ORANGE, 0.8)
        cuadrado2 = Square(side_length=1.5, color=PURPLE).set_fill(PURPLE, 0.8)
        
        cuadrado1.shift(LEFT * 3)
        cuadrado2.shift(RIGHT * 3)
        
        # Labels
        label1 = Text("ease_out_elastic", font_size=20).next_to(cuadrado1, DOWN, buff=1)
        label2 = Text("ease_in_out_elastic", font_size=20).next_to(cuadrado2, DOWN, buff=1)
        
        self.play(Write(titulo))
        self.play(Create(VGroup(cuadrado1, cuadrado2)))
        self.play(Write(VGroup(label1, label2)))
        self.wait(1)
        
        # Animación de escala con elastic
        self.play(
            cuadrado1.animate.scale(2),
            rate_func=ease_out_elastic,
            run_time=2
        )
        self.wait(0.5)
        
        self.play(
            cuadrado2.animate.scale(2),
            rate_func=ease_in_out_elastic,
            run_time=2
        )
        self.wait(1)
        
        # Restaurar
        self.play(
            cuadrado1.animate.scale(0.5),
            cuadrado2.animate.scale(0.5),
            rate_func=ease_out_elastic,
            run_time=2
        )
        self.wait(2)
````

## Rate Functions Personalizadas

Puedes crear tus propias rate functions. Una rate function es simplemente una función que:
- Recibe un valor `t` entre 0 y 1
- Retorna un valor (típicamente entre 0 y 1, pero puede salirse para efectos especiales)

```python
def mi_rate_func(t):
    # t va de 0 a 1
    # retorna el progreso de la animación
    return t ** 2  # Ejemplo: aceleración cuadrática
```

````python
%%manim -qm -v Warning RateFuncPersonalizada

class RateFuncPersonalizada(Scene):
    def construct(self):
        titulo = Text("Rate Functions Personalizadas", font_size=36)
        titulo.to_edge(UP)
        
        # Definir rate functions personalizadas
        def cuadratica(t):
            return t ** 2
        
        def cubica(t):
            return t ** 3
        
        def sinusoidal(t):
            return np.sin(t * PI / 2)
        
        def escalonada(t):
            # Movimiento en 5 pasos
            return np.floor(t * 5) / 5
        
        # Crear elementos
        circulo1 = Circle(radius=0.5, color=RED).set_fill(RED, 0.8)
        circulo2 = Circle(radius=0.5, color=GREEN).set_fill(GREEN, 0.8)
        circulo3 = Circle(radius=0.5, color=BLUE).set_fill(BLUE, 0.8)
        circulo4 = Circle(radius=0.5, color=YELLOW).set_fill(YELLOW, 0.8)
        
        # Posicionar
        circulos = VGroup(circulo1, circulo2, circulo3, circulo4)
        for i, c in enumerate(circulos):
            c.shift(LEFT * 5 + DOWN * (1.5 - i))
        
        # Labels
        labels = VGroup(
            Text("cuadrática (t²)", font_size=18),
            Text("cúbica (t³)", font_size=18),
            Text("sinusoidal", font_size=18),
            Text("escalonada", font_size=18)
        )
        for i, (label, circulo) in enumerate(zip(labels, circulos)):
            label.next_to(circulo, LEFT, buff=0.3)
        
        self.play(Write(titulo))
        self.play(Create(circulos), Write(labels))
        self.wait(1)
        
        # Animar con cada rate function
        self.play(
            circulo1.animate.shift(RIGHT * 10),
            rate_func=cuadratica,
            run_time=3
        )
        self.play(
            circulo2.animate.shift(RIGHT * 10),
            rate_func=cubica,
            run_time=3
        )
        self.play(
            circulo3.animate.shift(RIGHT * 10),
            rate_func=sinusoidal,
            run_time=3
        )
        self.play(
            circulo4.animate.shift(RIGHT * 10),
            rate_func=escalonada,
            run_time=3
        )
        
        self.wait(2)
````

## Visualización de Rate Functions

Es útil ver gráficamente cómo se comportan las diferentes rate functions:

````python
%%manim -qm -v Warning GraficaRateFunctions

class GraficaRateFunctions(Scene):
    def construct(self):
        titulo = Text("Gráficas de Rate Functions", font_size=36)
        titulo.to_edge(UP)
        
        # Crear ejes
        axes = Axes(
            x_range=[0, 1, 0.25],
            y_range=[0, 1.2, 0.25],
            x_length=5,
            y_length=4,
            axis_config={"include_tip": True, "include_numbers": True}
        )
        axes.shift(LEFT * 3)
        
        labels = axes.get_axis_labels(x_label="t", y_label="progreso")
        
        # Graficar diferentes rate functions
        grafica_linear = axes.plot(lambda t: t, color=WHITE, x_range=[0, 1])
        grafica_smooth = axes.plot(lambda t: smooth(t), color=BLUE, x_range=[0, 1])
        grafica_rush_into = axes.plot(lambda t: rush_into(t), color=RED, x_range=[0, 1])
        
        # Leyenda
        leyenda = VGroup(
            VGroup(Line(LEFT * 0.3, RIGHT * 0.3, color=WHITE), Text("linear", font_size=18)).arrange(RIGHT, buff=0.2),
            VGroup(Line(LEFT * 0.3, RIGHT * 0.3, color=BLUE), Text("smooth", font_size=18)).arrange(RIGHT, buff=0.2),
            VGroup(Line(LEFT * 0.3, RIGHT * 0.3, color=RED), Text("rush_into", font_size=18)).arrange(RIGHT, buff=0.2),
        ).arrange(DOWN, aligned_edge=LEFT)
        leyenda.shift(RIGHT * 4)
        
        self.play(Write(titulo))
        self.play(Create(axes), Write(labels))
        self.play(Create(grafica_linear), run_time=2)
        self.play(Create(grafica_smooth), run_time=2)
        self.play(Create(grafica_rush_into), run_time=2)
        self.play(Write(leyenda))
        self.wait(3)
````
