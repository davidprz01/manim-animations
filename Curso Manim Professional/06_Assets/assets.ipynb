## ImageMobject: Trabajando con Imágenes

`ImageMobject` permite cargar y manipular imágenes rasterizadas (PNG, JPG, GIF, etc.).

### Sintaxis Básica
```python
imagen = ImageMobject("ruta/a/imagen.png")
imagen.scale(2)  # Escalar
imagen.shift(UP)  # Posicionar
```

### Propiedades Importantes
- `.scale()`: Cambiar tamaño
- `.set_width()` / `.set_height()`: Establecer dimensiones específicas
- `.shift()`, `.move_to()`: Posicionar
- `.set_opacity()`: Cambiar transparencia

````python
%%manim -qm -v Warning EjemploImageMobject

class EjemploImageMobject(Scene):
    def construct(self):
        titulo = Text("ImageMobject - Ejemplo Conceptual", font_size=36)
        titulo.to_edge(UP)
        
        # Nota: Este ejemplo es conceptual
        # Para usar ImageMobject necesitas una imagen real
        # imagen = ImageMobject("assets/images/mi_imagen.png")
        
        # Simulación con un rectángulo que representa la imagen
        placeholder = Rectangle(width=4, height=3, color=BLUE)
        placeholder.set_fill(BLUE_E, opacity=0.5)
        
        texto_imagen = Text("Aquí iría tu imagen\nImageMobject('ruta.png')", font_size=20)
        texto_imagen.move_to(placeholder)
        
        grupo_imagen = VGroup(placeholder, texto_imagen)
        
        # Código de ejemplo
        codigo = Code(
            code='''# Cargar imagen
imagen = ImageMobject("logo.png")

# Manipular
imagen.scale(2)
imagen.shift(LEFT * 2)
imagen.set_opacity(0.8)

# Animar
self.play(FadeIn(imagen))''',
            language="python",
            font_size=18,
            background="window"
        )
        codigo.scale(0.8)
        codigo.shift(RIGHT * 3)
        
        self.play(Write(titulo))
        self.play(FadeIn(grupo_imagen))
        self.play(Create(codigo))
        self.wait(2)
        
        # Demostrar transformaciones
        self.play(grupo_imagen.animate.scale(0.7).shift(LEFT * 2))
        self.play(grupo_imagen.animate.set_opacity(0.5))
        self.play(grupo_imagen.animate.set_opacity(1))
        self.wait(2)
````

## SVGMobject: Trabajando con Vectores

`SVGMobject` carga archivos SVG y los convierte en objetos vectoriales de Manim. 

### Ventajas de SVG
✅ **Escalabilidad**: No pierden calidad al escalar  
✅ **Manipulación**: Puedes acceder a cada elemento del SVG  
✅ **Colores**: Puedes cambiar colores de partes individuales  
✅ **Animaciones**: Se pueden animar como cualquier VMobject

### Sintaxis
```python
svg = SVGMobject("ruta/archivo.svg")
svg.scale(2)
svg.set_color(RED)
```

````python
%%manim -qm -v Warning EjemploSVGMobject

class EjemploSVGMobject(Scene):
    def construct(self):
        titulo = Text("SVGMobject - Gráficos Vectoriales", font_size=36)
        titulo.to_edge(UP)
        
        # Nota: Este ejemplo es conceptual
        # Para usar SVGMobject necesitas un archivo SVG real
        # svg = SVGMobject("assets/svg/icono.svg")
        
        # Simulación: crear un "SVG" usando formas de Manim
        # Esto representa lo que podrías hacer con un SVG real
        
        icono_simulado = VGroup(
            Circle(radius=1, color=BLUE).set_fill(BLUE, 0.3),
            Triangle(color=WHITE).scale(0.5).set_fill(WHITE, 0.9),
        )
        
        label = Text("SVGMobject simulado", font_size=20)
        label.next_to(icono_simulado, DOWN)
        
        # Código de ejemplo
        codigo = Code(
            code='''# Cargar SVG
svg = SVGMobject("icono.svg")

# Cambiar colores
svg.set_color(BLUE)
svg[0].set_color(RED)  # Primer elemento

# Escalar y posicionar
svg.scale(2)
svg.to_edge(LEFT)

# Animar
self.play(DrawBorderThenFill(svg))''',
            language="python",
            font_size=18,
            background="window"
        )
        codigo.scale(0.75)
        codigo.shift(RIGHT * 3)
        
        self.play(Write(titulo))
        self.play(Create(icono_simulado), Write(label))
        self.play(Create(codigo))
        self.wait(2)
        
        # Demostrar manipulación de partes
        self.play(icono_simulado[0].animate.set_color(RED))
        self.play(icono_simulado[1].animate.set_color(YELLOW))
        self.play(icono_simulado.animate.scale(1.5).shift(LEFT * 2))
        self.wait(2)
````

## Accediendo a Elementos de un SVG

Los SVGs complejos contienen múltiples elementos. Puedes acceder a ellos como una lista:

````python
%%manim -qm -v Warning AccesoElementosSVG

class AccesoElementosSVG(Scene):
    def construct(self):
        titulo = Text("Acceso a Elementos de SVG", font_size=36)
        titulo.to_edge(UP)
        
        # Simulamos un SVG con múltiples partes
        # En un SVG real, cada path/shape es un elemento
        svg_simulado = VGroup(
            Circle(radius=0.5, color=RED).shift(LEFT * 2),
            Square(side_length=1, color=GREEN),
            Triangle(color=BLUE).scale(0.7).shift(RIGHT * 2),
            Star(n=5, color=YELLOW).scale(0.5).shift(RIGHT * 4),
        )
        
        # Labels para cada elemento
        for i, elem in enumerate(svg_simulado):
            label = Text(f"[{i}]", font_size=16)
            label.next_to(elem, DOWN)
            svg_simulado.add(label)
        
        info = Text("Un SVG es como una lista de elementos", font_size=24)
        info.to_edge(DOWN)
        
        self.play(Write(titulo))
        self.play(Create(svg_simulado[:4]))  # Solo las formas
        self.play(Write(info))
        self.wait(1)
        
        # Demostrar acceso individual
        self.play(svg_simulado[0].animate.scale(1.5).set_color(PINK))
        self.wait(0.5)
        
        self.play(svg_simulado[1].animate.rotate(PI/4).set_color(ORANGE))
        self.wait(0.5)
        
        self.play(svg_simulado[2].animate.shift(UP).set_color(PURPLE))
        self.wait(0.5)
        
        self.play(svg_simulado[3].animate.scale(2).set_color(WHITE))
        self.wait(2)
````

## Animaciones con Assets

Los assets (imágenes y SVGs) pueden animarse como cualquier otro Mobject:

````python
%%manim -qm -v Warning AnimacionesConAssets

class AnimacionesConAssets(Scene):
    def construct(self):
        titulo = Text("Animaciones Disponibles para Assets", font_size=36)
        titulo.to_edge(UP)
        
        # Crear "asset" simulado
        asset = VGroup(
            RoundedRectangle(corner_radius=0.3, width=2, height=2, color=BLUE),
            Text("LOGO", font_size=24, color=WHITE)
        )
        asset[1].move_to(asset[0])
        asset.set_fill(BLUE_E, opacity=0.8)
        
        info_animaciones = VGroup(
            Text("• FadeIn / FadeOut", font_size=20),
            Text("• GrowFromCenter", font_size=20),
            Text("• DrawBorderThenFill (SVG)", font_size=20),
            Text("• SpinInFromNothing", font_size=20),
            Text("• Transform", font_size=20),
        ).arrange(DOWN, aligned_edge=LEFT)
        info_animaciones.to_edge(RIGHT)
        
        self.play(Write(titulo))
        self.play(Write(info_animaciones))
        self.wait(1)
        
        # FadeIn
        self.play(FadeIn(asset))
        self.wait(0.5)
        self.play(FadeOut(asset))
        
        # GrowFromCenter
        self.play(GrowFromCenter(asset))
        self.wait(0.5)
        self.play(FadeOut(asset))
        
        # SpinInFromNothing
        self.play(SpinInFromNothing(asset))
        self.wait(0.5)
        
        # Transformaciones con .animate
        self.play(
            asset.animate.scale(1.5).shift(LEFT * 2).rotate(PI/6),
            run_time=2
        )
        self.wait(0.5)
        
        self.play(
            asset.animate.set_opacity(0.5),
            run_time=1
        )
        self.play(
            asset.animate.set_opacity(1),
            run_time=1
        )
        self.wait(2)
````

## Ejemplo Práctico: Galería de Imágenes

Veamos cómo crear una presentación tipo galería:

````python
%%manim -qm -v Warning GaleriaImagenes

class GaleriaImagenes(Scene):
    def construct(self):
        titulo = Text("Galería de Imágenes", font_size=40)
        titulo.to_edge(UP)
        
        # Crear "marcos" de imagen simulados
        colores = [RED, GREEN, BLUE, YELLOW, PURPLE]
        marcos = VGroup()
        
        for i, color in enumerate(colores):
            marco = VGroup(
                Rectangle(width=2.5, height=2, color=WHITE, stroke_width=3),
                Rectangle(width=2.3, height=1.8, color=color).set_fill(color, 0.7),
                Text(f"Imagen {i+1}", font_size=16, color=WHITE)
            )
            marco[2].move_to(marco[1])
            marcos.add(marco)
        
        marcos.arrange(RIGHT, buff=0.5)
        marcos.scale(0.7)
        
        self.play(Write(titulo))
        
        # Animación de entrada uno por uno
        for marco in marcos:
            self.play(FadeIn(marco, shift=UP), run_time=0.5)
        
        self.wait(1)
        
        # Destacar una imagen
        imagen_destacada = marcos[2].copy()
        self.play(
            imagen_destacada.animate.scale(2).move_to(ORIGIN),
            marcos.animate.set_opacity(0.3)
        )
        self.wait(1)
        
        # Volver a galería
        self.play(
            FadeOut(imagen_destacada),
            marcos.animate.set_opacity(1)
        )
        self.wait(1)
        
        # Transición de slide
        self.play(marcos.animate.shift(LEFT * 3))
        self.wait(2)
````

## Consejos para Trabajar con Assets

### Imágenes (ImageMobject)
```python
# Cargar imagen
img = ImageMobject("imagen.png")

# Establecer dimensiones específicas
img.set_width(4)  # Ancho de 4 unidades Manim
img.set_height(3)  # Alto de 3 unidades Manim

# Mantener aspect ratio
img.scale_to_fit_width(4)
img.scale_to_fit_height(3)

# Transparencia
img.set_opacity(0.8)
```

### SVGs (SVGMobject)
```python
# Cargar SVG
svg = SVGMobject("icono.svg")

# Cambiar color de todo el SVG
svg.set_color(BLUE)

# Cambiar color de elementos específicos
svg[0].set_color(RED)
svg[1].set_fill(GREEN, opacity=0.5)

# Animación especial para SVG
self.play(DrawBorderThenFill(svg))
self.play(Write(svg))  # También funciona
```

## Ejemplo: Logo Animado con DrawBorderThenFill

````python
%%manim -qm -v Warning LogoAnimado

class LogoAnimado(Scene):
    def construct(self):
        # Crear un "logo" con formas de Manim
        # En la práctica, cargarías un SVG
        
        # Círculo exterior
        circulo = Circle(radius=2, color=BLUE)
        circulo.set_stroke(width=8)
        
        # Triángulo interior (play button)
        triangulo = Triangle(color=WHITE)
        triangulo.scale(0.8)
        triangulo.rotate(-PI/2)
        triangulo.set_fill(WHITE, opacity=1)
        
        # Texto
        texto = Text("PLAY", font_size=20, color=WHITE)
        texto.next_to(triangulo, DOWN, buff=0.3)
        
        logo = VGroup(circulo, triangulo, texto)
        
        # Animación estilo DrawBorderThenFill
        self.play(Create(circulo), run_time=2)
        self.play(DrawBorderThenFill(triangulo), run_time=1.5)
        self.play(Write(texto))
        self.wait(1)
        
        # Efecto de "pulso"
        self.play(
            logo.animate.scale(1.1),
            rate_func=there_and_back,
            run_time=0.5
        )
        self.play(
            logo.animate.scale(1.1),
            rate_func=there_and_back,
            run_time=0.5
        )
        
        # Rotación sutil
        self.play(Rotate(logo, angle=2*PI), run_time=2)
        self.wait(2)
````
