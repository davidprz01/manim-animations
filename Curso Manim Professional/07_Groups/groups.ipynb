````python
from manim import *
````

## ¿Por qué usar Groups?

Los grupos permiten:
- **Mover múltiples objetos** como una unidad
- **Aplicar transformaciones** a todos los elementos simultáneamente
- **Organizar automáticamente** con `arrange()`
- **Acceder a elementos** individualmente con índices
- **Animar el grupo completo** con una sola animación

### VGroup vs Group
| VGroup | Group |
|--------|-------|
| Solo VMobjects (vectoriales) | Cualquier Mobject |
| Circle, Square, Text, etc. | Incluye ImageMobject |
| Más métodos de manipulación | Más genérico |
| **Recomendado para la mayoría de casos** | Usar cuando hay imágenes |

## Creación Básica de VGroup

````python
%%manim -qm -v Warning CreacionVGroup

class CreacionVGroup(Scene):
    def construct(self):
        titulo = Text("Creación de VGroup", font_size=40)
        titulo.to_edge(UP)
        
        # Forma 1: Pasar elementos al constructor
        grupo1 = VGroup(
            Circle(radius=0.5, color=RED),
            Square(side_length=1, color=GREEN),
            Triangle(color=BLUE).scale(0.7)
        )
        
        # Forma 2: Crear vacío y agregar con .add()
        grupo2 = VGroup()
        grupo2.add(Circle(radius=0.5, color=YELLOW))
        grupo2.add(Square(side_length=1, color=PURPLE))
        grupo2.add(Triangle(color=ORANGE).scale(0.7))
        
        # Posicionar grupos
        grupo1.shift(LEFT * 3)
        grupo2.shift(RIGHT * 3)
        
        # Labels
        label1 = Text("VGroup(a, b, c)", font_size=20).next_to(grupo1, DOWN)
        label2 = Text("grupo.add(x)", font_size=20).next_to(grupo2, DOWN)
        
        self.play(Write(titulo))
        self.play(Create(grupo1), Create(grupo2))
        self.play(Write(label1), Write(label2))
        self.wait(2)
        
        # Demostrar que se mueven como unidad
        self.play(
            grupo1.animate.shift(UP),
            grupo2.animate.shift(DOWN)
        )
        self.wait(2)
````

## Método arrange(): Organización Automática

`arrange()` distribuye los elementos del grupo automáticamente:

```python
grupo.arrange(direction, buff=0.25)
```

- `direction`: UP, DOWN, LEFT, RIGHT
- `buff`: Espacio entre elementos (default: 0.25)

````python
%%manim -qm -v Warning MetodoArrange

class MetodoArrange(Scene):
    def construct(self):
        titulo = Text("Método arrange()", font_size=40)
        titulo.to_edge(UP)
        
        # Crear formas
        formas = VGroup(
            Circle(radius=0.5, color=RED).set_fill(RED, 0.5),
            Square(side_length=1, color=GREEN).set_fill(GREEN, 0.5),
            Triangle(color=BLUE).scale(0.7).set_fill(BLUE, 0.5),
            Star(n=5, color=YELLOW).scale(0.5).set_fill(YELLOW, 0.5),
        )
        
        self.play(Write(titulo))
        self.play(Create(formas))
        self.wait(1)
        
        # arrange horizontal (RIGHT es default)
        info = Text("arrange(RIGHT, buff=0.5)", font_size=24).to_edge(DOWN)
        self.play(Write(info))
        self.play(formas.animate.arrange(RIGHT, buff=0.5))
        self.wait(1)
        
        # arrange vertical
        self.play(Transform(info, Text("arrange(DOWN, buff=0.3)", font_size=24).to_edge(DOWN)))
        self.play(formas.animate.arrange(DOWN, buff=0.3))
        self.wait(1)
        
        # arrange con alineación
        self.play(Transform(info, Text("arrange(RIGHT, aligned_edge=DOWN)", font_size=24).to_edge(DOWN)))
        self.play(formas.animate.arrange(RIGHT, aligned_edge=DOWN, buff=0.5))
        self.wait(1)
        
        # arrange con alineación superior
        self.play(Transform(info, Text("arrange(RIGHT, aligned_edge=UP)", font_size=24).to_edge(DOWN)))
        self.play(formas.animate.arrange(RIGHT, aligned_edge=UP, buff=0.5))
        self.wait(2)
````

## Método arrange_in_grid(): Organización en Cuadrícula

Perfecto para crear matrices, tableros, o cualquier disposición en filas y columnas:

````python
%%manim -qm -v Warning ArrangeInGrid

class ArrangeInGrid(Scene):
    def construct(self):
        titulo = Text("arrange_in_grid()", font_size=40)
        titulo.to_edge(UP)
        
        # Crear 12 cuadrados
        cuadrados = VGroup(*[
            Square(side_length=0.8, color=interpolate_color(BLUE, RED, i/11))
            .set_fill(interpolate_color(BLUE, RED, i/11), 0.7)
            for i in range(12)
        ])
        
        self.play(Write(titulo))
        self.play(Create(cuadrados))
        self.wait(1)
        
        # Grid 3x4
        info = Text("arrange_in_grid(rows=3, cols=4)", font_size=22).to_edge(DOWN)
        self.play(Write(info))
        self.play(cuadrados.animate.arrange_in_grid(rows=3, cols=4, buff=0.3))
        self.wait(1)
        
        # Grid 4x3
        self.play(Transform(info, Text("arrange_in_grid(rows=4, cols=3)", font_size=22).to_edge(DOWN)))
        self.play(cuadrados.animate.arrange_in_grid(rows=4, cols=3, buff=0.3))
        self.wait(1)
        
        # Grid 2x6
        self.play(Transform(info, Text("arrange_in_grid(rows=2, cols=6)", font_size=22).to_edge(DOWN)))
        self.play(cuadrados.animate.arrange_in_grid(rows=2, cols=6, buff=0.2))
        self.wait(1)
        
        # Grid 6x2 con más espacio
        self.play(Transform(info, Text("arrange_in_grid(rows=6, cols=2, buff=0.5)", font_size=22).to_edge(DOWN)))
        self.play(cuadrados.animate.arrange_in_grid(rows=6, cols=2, buff=0.5))
        self.wait(2)
````

## Acceso a Elementos Individuales

Puedes acceder a elementos del grupo usando índices, como una lista de Python:

````python
%%manim -qm -v Warning AccesoElementos

class AccesoElementos(Scene):
    def construct(self):
        titulo = Text("Acceso a Elementos", font_size=40)
        titulo.to_edge(UP)
        
        # Crear grupo
        grupo = VGroup(
            Circle(radius=0.6, color=RED).set_fill(RED, 0.7),
            Square(side_length=1.2, color=GREEN).set_fill(GREEN, 0.7),
            Triangle(color=BLUE).scale(0.8).set_fill(BLUE, 0.7),
            Star(n=5, color=YELLOW).scale(0.5).set_fill(YELLOW, 0.7),
        )
        grupo.arrange(RIGHT, buff=1)
        
        # Labels con índices
        for i, elem in enumerate(grupo):
            label = Text(f"[{i}]", font_size=20).next_to(elem, DOWN)
            grupo.add(label)
        
        self.play(Write(titulo))
        self.play(Create(grupo[:4]))  # Solo las formas
        self.play(Write(grupo[4:]))   # Los labels
        self.wait(1)
        
        # Acceso individual
        info = Text("grupo[0] - Primer elemento", font_size=24).to_edge(DOWN)
        self.play(Write(info))
        self.play(grupo[0].animate.scale(1.5).set_color(WHITE))
        self.wait(0.5)
        
        self.play(Transform(info, Text("grupo[1] - Segundo elemento", font_size=24).to_edge(DOWN)))
        self.play(grupo[1].animate.rotate(PI/4))
        self.wait(0.5)
        
        self.play(Transform(info, Text("grupo[-1] - Último elemento", font_size=24).to_edge(DOWN)))
        self.play(grupo[3].animate.scale(2))
        self.wait(0.5)
        
        # Slicing
        self.play(Transform(info, Text("grupo[1:3] - Slice de elementos", font_size=24).to_edge(DOWN)))
        self.play(grupo[1:3].animate.set_color(PURPLE))
        self.wait(2)
````

## Transformaciones en Grupo

Todas las transformaciones aplicadas a un grupo afectan a todos sus elementos:

````python
%%manim -qm -v Warning TransformacionesGrupo

class TransformacionesGrupo(Scene):
    def construct(self):
        titulo = Text("Transformaciones en Grupo", font_size=36)
        titulo.to_edge(UP)
        
        # Crear grupo
        grupo = VGroup(
            Circle(radius=0.5, color=RED).set_fill(RED, 0.7),
            Square(side_length=1, color=GREEN).set_fill(GREEN, 0.7),
            Triangle(color=BLUE).scale(0.7).set_fill(BLUE, 0.7),
        )
        grupo.arrange(RIGHT, buff=0.5)
        
        self.play(Write(titulo))
        self.play(Create(grupo))
        self.wait(1)
        
        # Escalar grupo
        info = Text("grupo.scale(1.5)", font_size=24).to_edge(DOWN)
        self.play(Write(info))
        self.play(grupo.animate.scale(1.5))
        self.wait(0.5)
        
        # Rotar grupo
        self.play(Transform(info, Text("grupo.rotate(PI/4)", font_size=24).to_edge(DOWN)))
        self.play(Rotate(grupo, angle=PI/4))
        self.wait(0.5)
        
        # Mover grupo
        self.play(Transform(info, Text("grupo.shift(LEFT * 2)", font_size=24).to_edge(DOWN)))
        self.play(grupo.animate.shift(LEFT * 2))
        self.wait(0.5)
        
        # Cambiar color de todo el grupo
        self.play(Transform(info, Text("grupo.set_color(YELLOW)", font_size=24).to_edge(DOWN)))
        self.play(grupo.animate.set_color(YELLOW))
        self.wait(0.5)
        
        # Opacidad del grupo
        self.play(Transform(info, Text("grupo.set_opacity(0.3)", font_size=24).to_edge(DOWN)))
        self.play(grupo.animate.set_opacity(0.3))
        self.play(grupo.animate.set_opacity(1))
        self.wait(2)
````

## Grupos Anidados

Puedes crear grupos dentro de grupos para estructuras más complejas:

````python
%%manim -qm -v Warning GruposAnidados

class GruposAnidados(Scene):
    def construct(self):
        titulo = Text("Grupos Anidados", font_size=40)
        titulo.to_edge(UP)
        
        # Crear subgrupos
        grupo_circulos = VGroup(
            Circle(radius=0.4, color=RED).set_fill(RED, 0.7),
            Circle(radius=0.4, color=RED).set_fill(RED, 0.7),
            Circle(radius=0.4, color=RED).set_fill(RED, 0.7),
        )
        grupo_circulos.arrange(DOWN, buff=0.2)
        
        grupo_cuadrados = VGroup(
            Square(side_length=0.8, color=BLUE).set_fill(BLUE, 0.7),
            Square(side_length=0.8, color=BLUE).set_fill(BLUE, 0.7),
            Square(side_length=0.8, color=BLUE).set_fill(BLUE, 0.7),
        )
        grupo_cuadrados.arrange(DOWN, buff=0.2)
        
        grupo_triangulos = VGroup(
            Triangle(color=GREEN).scale(0.5).set_fill(GREEN, 0.7),
            Triangle(color=GREEN).scale(0.5).set_fill(GREEN, 0.7),
            Triangle(color=GREEN).scale(0.5).set_fill(GREEN, 0.7),
        )
        grupo_triangulos.arrange(DOWN, buff=0.2)
        
        # Grupo principal
        grupo_principal = VGroup(grupo_circulos, grupo_cuadrados, grupo_triangulos)
        grupo_principal.arrange(RIGHT, buff=1)
        
        # Labels
        label_c = Text("grupo[0]", font_size=16).next_to(grupo_circulos, DOWN)
        label_s = Text("grupo[1]", font_size=16).next_to(grupo_cuadrados, DOWN)
        label_t = Text("grupo[2]", font_size=16).next_to(grupo_triangulos, DOWN)
        
        self.play(Write(titulo))
        self.play(Create(grupo_principal))
        self.play(Write(VGroup(label_c, label_s, label_t)))
        self.wait(1)
        
        # Manipular subgrupo
        info = Text("grupo[0].set_color(YELLOW) - Subgrupo completo", font_size=20).to_edge(DOWN)
        self.play(Write(info))
        self.play(grupo_principal[0].animate.set_color(YELLOW))
        self.wait(0.5)
        
        # Elemento específico en subgrupo
        self.play(Transform(info, Text("grupo[1][1].scale(1.5) - Elemento del subgrupo", font_size=20).to_edge(DOWN)))
        self.play(grupo_principal[1][1].animate.scale(1.5))
        self.wait(0.5)
        
        # Mover subgrupo
        self.play(Transform(info, Text("grupo[2].shift(UP) - Mover subgrupo", font_size=20).to_edge(DOWN)))
        self.play(grupo_principal[2].animate.shift(UP))
        self.wait(2)
````

## Ejemplo Práctico: Creación de Tarjetas

````python
%%manim -qm -v Warning CreacionTarjetas

class CreacionTarjetas(Scene):
    def construct(self):
        titulo = Text("Ejemplo: Tarjetas con VGroup", font_size=36)
        titulo.to_edge(UP)
        
        def crear_tarjeta(icono, titulo_text, descripcion, color):
            """Función para crear una tarjeta estructurada"""
            # Fondo de tarjeta
            fondo = RoundedRectangle(
                corner_radius=0.2,
                width=3,
                height=4,
                color=color
            ).set_fill(color, 0.2)
            
            # Icono
            icono_mob = icono.copy().scale(0.8)
            icono_mob.set_color(color)
            
            # Título
            titulo_mob = Text(titulo_text, font_size=22, color=WHITE)
            
            # Descripción
            desc_mob = Text(descripcion, font_size=14, color=GRAY)
            
            # Organizar verticalmente
            contenido = VGroup(icono_mob, titulo_mob, desc_mob)
            contenido.arrange(DOWN, buff=0.3)
            contenido.move_to(fondo)
            
            return VGroup(fondo, contenido)
        
        # Crear tarjetas
        tarjeta1 = crear_tarjeta(Circle(radius=0.5), "Círculos", "Formas redondas", BLUE)
        tarjeta2 = crear_tarjeta(Square(side_length=1), "Cuadrados", "Formas cuadradas", GREEN)
        tarjeta3 = crear_tarjeta(Triangle().scale(0.7), "Triángulos", "Formas triangulares", RED)
        
        # Agrupar tarjetas
        tarjetas = VGroup(tarjeta1, tarjeta2, tarjeta3)
        tarjetas.arrange(RIGHT, buff=0.5)
        tarjetas.scale(0.85)
        
        self.play(Write(titulo))
        
        # Animación de entrada
        for tarjeta in tarjetas:
            self.play(FadeIn(tarjeta, shift=UP), run_time=0.5)
        
        self.wait(1)
        
        # Destacar una tarjeta
        self.play(
            tarjeta2.animate.scale(1.15),
            tarjeta1.animate.set_opacity(0.5),
            tarjeta3.animate.set_opacity(0.5)
        )
        self.wait(1)
        
        # Restaurar
        self.play(
            tarjeta2.animate.scale(1/1.15),
            tarjeta1.animate.set_opacity(1),
            tarjeta3.animate.set_opacity(1)
        )
        self.wait(2)
````

## Iteración sobre Grupos

Puedes iterar sobre los elementos de un grupo con un bucle for:

````python
%%manim -qm -v Warning IteracionGrupos

class IteracionGrupos(Scene):
    def construct(self):
        titulo = Text("Iteración sobre Grupos", font_size=40)
        titulo.to_edge(UP)
        
        # Crear grupo de círculos
        circulos = VGroup(*[
            Circle(radius=0.5, color=WHITE).set_fill(WHITE, 0.5)
            for _ in range(8)
        ])
        circulos.arrange(RIGHT, buff=0.3)
        circulos.scale(0.8)
        
        self.play(Write(titulo))
        self.play(Create(circulos))
        self.wait(1)
        
        # Iterar y cambiar colores uno por uno
        colores = [RED, ORANGE, YELLOW, GREEN, BLUE, PURPLE, PINK, TEAL]
        
        info = Text("for i, circulo in enumerate(grupo):", font_size=20).to_edge(DOWN)
        self.play(Write(info))
        
        for i, circulo in enumerate(circulos):
            self.play(
                circulo.animate.set_color(colores[i]).set_fill(colores[i], 0.8),
                run_time=0.3
            )
        
        self.wait(1)
        
        # Efecto de onda
        self.play(Transform(info, Text("Efecto de onda con iteración", font_size=20).to_edge(DOWN)))
        
        for _ in range(2):
            for circulo in circulos:
                self.play(
                    circulo.animate.scale(1.3),
                    run_time=0.1
                )
                self.play(
                    circulo.animate.scale(1/1.3),
                    run_time=0.1
                )
        
        self.wait(2)
````

### Conceptos Clave de esta Lección
✅ **VGroup**: Agrupa VMobjects vectoriales  
✅ **Group**: Agrupa cualquier Mobject (incluidas imágenes)  
✅ **arrange(direction, buff)**: Organiza elementos en línea  
✅ **arrange_in_grid(rows, cols)**: Organiza en cuadrícula  
✅ **aligned_edge**: Alinea elementos por un borde (UP, DOWN, LEFT, RIGHT)  
✅ **Acceso por índice**: grupo[0], grupo[1], grupo[-1]  
✅ **Slicing**: grupo[1:3], grupo[:2]  
✅ **Grupos anidados**: grupo[i][j] para acceder a subelementos  
✅ **Iteración**: for elemento in grupo

### Tabla de Métodos Importantes

| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| `VGroup(a, b, c)` | Crear grupo | `VGroup(c1, c2, c3)` |
| `.add(mob)` | Agregar elemento | `grupo.add(circulo)` |
| `.remove(mob)` | Quitar elemento | `grupo.remove(circulo)` |
| `.arrange(dir)` | Organizar en línea | `grupo.arrange(RIGHT)` |
| `.arrange_in_grid()` | Organizar en grid | `grupo.arrange_in_grid(3, 4)` |
| `grupo[i]` | Acceder elemento | `grupo[0].set_color(RED)` |
| `len(grupo)` | Número de elementos | `n = len(grupo)` |
