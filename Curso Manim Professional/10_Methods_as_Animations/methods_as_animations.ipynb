## 1. 🎭 Introducción a .animate

El atributo `.animate` es una de las características más poderosas de Manim.
Permite convertir **cualquier método de un mobject** en una animación.

### Sintaxis básica:
```python
# Sin animación (instantáneo):
circulo.shift(RIGHT)

# Con animación:
self.play(circulo.animate.shift(RIGHT))
```

### ¿Cómo funciona?
1. `.animate` crea un "proxy" del objeto
2. El método aplicado define el estado final
3. `self.play()` interpola entre el estado actual y el final

### Ventajas:
- ✅ Código más limpio y legible
- ✅ Fácil de encadenar múltiples métodos
- ✅ Compatible con todos los métodos de mobjects

````python
%%manim -qm -v WARNING AnimateIntro

class AnimateIntro(Scene):
    def construct(self):
        titulo = Text(".animate - El Método Mágico", font_size=36).to_edge(UP)
        self.play(Write(titulo))
        
        # Sin .animate (instantáneo)
        circulo = Circle(radius=1, color=BLUE, fill_opacity=0.7)
        label_sin = Text("Sin .animate", font_size=20, color=RED).shift(UP * 2 + LEFT * 3)
        
        self.play(Create(circulo), Write(label_sin))
        circulo.shift(RIGHT * 2)  # ¡Instantáneo! No se ve la animación
        self.wait(0.5)
        
        # Con .animate
        circulo2 = Circle(radius=1, color=GREEN, fill_opacity=0.7)
        circulo2.shift(LEFT * 5)
        label_con = Text("Con .animate", font_size=20, color=GREEN).shift(DOWN * 2 + LEFT * 3)
        
        self.play(Create(circulo2), Write(label_con))
        self.play(circulo2.animate.shift(RIGHT * 2))  # ¡Animado!
        
        # Demostrar la diferencia
        nota = Text("El círculo azul se movió instantáneamente", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(nota))
        
        self.wait()
````

## 2. Métodos de Posición con .animate

Los métodos de posición permiten mover objetos en la escena. Con `.animate`, estos movimientos se convierten en animaciones suaves.

### Métodos principales:
- **`shift(vector)`**: Desplazamiento relativo
- **`move_to(posición)`**: Movimiento a posición absoluta
- **`next_to(objeto, dirección)`**: Posicionarse junto a otro objeto
- **`to_edge(dirección)`**: Mover al borde de la pantalla
- **`to_corner(esquina)`**: Mover a una esquina

````python
%%manim -qm -v WARNING AnimateShift

class AnimateShift(Scene):
    def construct(self):
        titulo = Text("shift() - Desplazamiento Relativo", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objetos
        cuadrado = Square(side_length=1.5, color=BLUE, fill_opacity=0.7)
        punto_inicio = Dot(ORIGIN, color=YELLOW).scale(0.5)
        label = Text("Inicio", font_size=16).next_to(punto_inicio, DOWN)
        
        self.play(Create(cuadrado), Create(punto_inicio), Write(label))
        
        # Desplazamientos con shift
        direcciones = [
            (RIGHT * 3, "RIGHT * 3"),
            (UP * 2, "UP * 2"),
            (LEFT * 3, "LEFT * 3"),
            (DOWN * 2 + RIGHT * 3, "DOWN*2 + RIGHT*3")
        ]
        
        for vector, nombre in direcciones:
            info = Text(f".animate.shift({nombre})", font_size=18, color=YELLOW)
            info.to_edge(DOWN)
            self.play(Write(info))
            self.play(cuadrado.animate.shift(vector))
            self.play(FadeOut(info))
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateMoveTo

class AnimateMoveTo(Scene):
    def construct(self):
        titulo = Text("move_to() - Posición Absoluta", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Grid de referencia
        grid = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-4, 4, 1],
            background_line_style={"stroke_opacity": 0.3}
        )
        self.play(Create(grid, run_time=1))
        
        # Objeto a mover
        circulo = Circle(radius=0.5, color=RED, fill_opacity=0.8)
        pos_label = always_redraw(
            lambda: Text(f"({circulo.get_center()[0]:.1f}, {circulo.get_center()[1]:.1f})", 
                        font_size=16, color=YELLOW).next_to(circulo, UP)
        )
        
        self.play(Create(circulo), Write(pos_label))
        
        # Movimientos a posiciones absolutas
        posiciones = [
            (np.array([3, 2, 0]), "move_to([3, 2, 0])"),
            (np.array([-3, 2, 0]), "move_to([-3, 2, 0])"),
            (np.array([-3, -2, 0]), "move_to([-3, -2, 0])"),
            (np.array([0, 0, 0]), "move_to(ORIGIN)")
        ]
        
        for pos, texto in posiciones:
            info = Text(f".animate.{texto}", font_size=18, color=GREEN).to_edge(DOWN)
            self.play(Write(info))
            self.play(circulo.animate.move_to(pos))
            self.play(FadeOut(info))
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateNextTo

class AnimateNextTo(Scene):
    def construct(self):
        titulo = Text("next_to() - Posición Relativa a Otro Objeto", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Objeto central de referencia
        centro = Square(side_length=2, color=BLUE, fill_opacity=0.7)
        label_centro = Text("Referencia", font_size=16, color=WHITE).move_to(centro)
        self.play(Create(centro), Write(label_centro))
        
        # Objeto que se moverá
        circulo = Circle(radius=0.4, color=RED, fill_opacity=0.8)
        circulo.next_to(centro, UP)
        self.play(Create(circulo))
        
        # Demostrar diferentes posiciones
        direcciones = [
            (RIGHT, "RIGHT"),
            (DOWN, "DOWN"),
            (LEFT, "LEFT"),
            (UR, "UR (Upper Right)"),
            (DL, "DL (Down Left)")
        ]
        
        for direccion, nombre in direcciones:
            info = Text(f".animate.next_to(cuadrado, {nombre})", 
                       font_size=16, color=YELLOW).to_edge(DOWN)
            self.play(Write(info))
            self.play(circulo.animate.next_to(centro, direccion, buff=0.2))
            self.play(FadeOut(info))
        
        # Demostrar con buff
        info_buff = Text("buff controla la separación", 
                        font_size=18, color=GREEN).to_edge(DOWN)
        self.play(Write(info_buff))
        self.play(circulo.animate.next_to(centro, RIGHT, buff=1))
        self.play(circulo.animate.next_to(centro, RIGHT, buff=0.1))
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateToEdgeCorner

class AnimateToEdgeCorner(Scene):
    def construct(self):
        titulo = Text("to_edge() y to_corner()", font_size=30).to_edge(UP, buff=0.3)
        self.play(Write(titulo))
        
        # Objeto para to_edge
        cuadrado = Square(side_length=1, color=BLUE, fill_opacity=0.7)
        self.play(Create(cuadrado))
        
        # to_edge demo
        bordes = [(LEFT, "LEFT"), (DOWN, "DOWN"), (RIGHT, "RIGHT")]
        
        for borde, nombre in bordes:
            info = Text(f".animate.to_edge({nombre})", 
                       font_size=18, color=YELLOW).shift(UP * 2)
            self.play(Write(info))
            self.play(cuadrado.animate.to_edge(borde))
            self.play(FadeOut(info))
        
        self.play(FadeOut(cuadrado))
        
        # to_corner demo
        triangulo = Triangle(color=GREEN, fill_opacity=0.7).scale(0.7)
        self.play(Create(triangulo))
        
        esquinas = [
            (UL, "UL (Upper Left)"),
            (UR, "UR (Upper Right)"),
            (DR, "DR (Down Right)"),
            (DL, "DL (Down Left)")
        ]
        
        for esquina, nombre in esquinas:
            info = Text(f".animate.to_corner({nombre})", 
                       font_size=16, color=YELLOW).shift(DOWN)
            self.play(Write(info))
            self.play(triangulo.animate.to_corner(esquina))
            self.play(FadeOut(info))
        
        self.wait()
````

## 3. Métodos de Escala con .animate

Los métodos de escala permiten cambiar el tamaño de los objetos de forma animada.

### Métodos principales:
- **`scale(factor)`**: Escalar uniformemente
- **`scale_to_fit_width(ancho)`**: Escalar para ajustar al ancho
- **`scale_to_fit_height(alto)`**: Escalar para ajustar a la altura
- **`stretch(factor, dim)`**: Estirar en una dimensión específica
- **`stretch_to_fit_width/height`**: Estirar para ajustar

````python
%%manim -qm -v WARNING AnimateScale

class AnimateScale(Scene):
    def construct(self):
        titulo = Text("scale() - Escalado Uniforme", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear varios objetos
        circulo = Circle(radius=0.5, color=BLUE, fill_opacity=0.7)
        cuadrado = Square(side_length=1, color=RED, fill_opacity=0.7)
        triangulo = Triangle(color=GREEN, fill_opacity=0.7).scale(0.5)
        
        objetos = VGroup(circulo, cuadrado, triangulo).arrange(RIGHT, buff=2)
        
        self.play(*[Create(obj) for obj in objetos])
        
        # Escalar hacia arriba
        info1 = Text(".animate.scale(2) - Duplicar tamaño", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(
            circulo.animate.scale(2),
            cuadrado.animate.scale(2),
            triangulo.animate.scale(2)
        )
        self.wait(0.5)
        
        # Escalar hacia abajo
        self.play(FadeOut(info1))
        info2 = Text(".animate.scale(0.3) - Reducir tamaño", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(
            circulo.animate.scale(0.3),
            cuadrado.animate.scale(0.3),
            triangulo.animate.scale(0.3)
        )
        self.wait(0.5)
        
        # Volver al tamaño original (aproximado)
        self.play(FadeOut(info2))
        info3 = Text(".animate.scale(1.67) - Restaurar", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(
            circulo.animate.scale(1.67),
            cuadrado.animate.scale(1.67),
            triangulo.animate.scale(1.67)
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateStretch

class AnimateStretch(Scene):
    def construct(self):
        titulo = Text("stretch() - Estirar en una Dimensión", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear círculo original
        circulo = Circle(radius=1, color=BLUE, fill_opacity=0.7)
        self.play(Create(circulo))
        
        # stretch en X (dim=0)
        info1 = Text(".animate.stretch(2, dim=0) - Estirar horizontalmente", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(circulo.animate.stretch(2, dim=0))
        self.wait(0.5)
        
        # stretch en Y (dim=1)
        self.play(FadeOut(info1))
        info2 = Text(".animate.stretch(0.5, dim=1) - Comprimir verticalmente", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(circulo.animate.stretch(0.5, dim=1))
        self.wait(0.5)
        
        self.play(FadeOut(circulo), FadeOut(info2))
        
        # Comparación scale_to_fit
        cuadrado = Square(side_length=2, color=GREEN, fill_opacity=0.7)
        self.play(Create(cuadrado))
        
        info3 = Text(".animate.scale_to_fit_width(5)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(cuadrado.animate.scale_to_fit_width(5))
        self.wait(0.5)
        
        self.play(FadeOut(info3))
        info4 = Text(".animate.scale_to_fit_height(1)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info4))
        self.play(cuadrado.animate.scale_to_fit_height(1))
        
        self.wait()
````

## 4. Métodos de Rotación con .animate

Los métodos de rotación permiten girar objetos alrededor de diferentes ejes y puntos.

### Métodos principales:
- **`rotate(ángulo)`**: Rotar en radianes (usar PI, TAU)
- **`rotate(ángulo, about_point=punto)`**: Rotar alrededor de un punto
- **`rotate(ángulo, axis=eje)`**: Rotar alrededor de un eje 3D
- **`flip(eje)`**: Voltear/reflejar respecto a un eje

### Constantes útiles:
- `PI` = 3.14159... (180°)
- `TAU` = 2*PI (360°)
- `DEGREES` - para convertir grados a radianes

````python
%%manim -qm -v WARNING AnimateRotate

class AnimateRotate(Scene):
    def construct(self):
        titulo = Text("rotate() - Rotación de Objetos", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objeto con orientación clara
        flecha = Arrow(ORIGIN, RIGHT * 2, color=BLUE, buff=0)
        label = Text("→", font_size=24, color=YELLOW).next_to(flecha, UP)
        self.play(Create(flecha))
        
        # Rotación básica
        info1 = Text(".animate.rotate(PI/2) - 90°", font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(flecha.animate.rotate(PI/2))
        self.wait(0.5)
        
        # Más rotaciones
        self.play(FadeOut(info1))
        info2 = Text(".animate.rotate(PI) - 180°", font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(flecha.animate.rotate(PI))
        self.wait(0.5)
        
        # Rotación usando DEGREES
        self.play(FadeOut(info2))
        info3 = Text(".animate.rotate(45*DEGREES)", font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(flecha.animate.rotate(45*DEGREES))
        self.wait(0.5)
        
        self.play(FadeOut(flecha), FadeOut(info3))
        
        # Rotación alrededor de un punto
        cuadrado = Square(side_length=1, color=GREEN, fill_opacity=0.7)
        cuadrado.shift(LEFT * 2)
        punto = Dot(ORIGIN, color=RED)
        label_punto = Text("Punto de rotación", font_size=14).next_to(punto, DOWN)
        
        self.play(Create(cuadrado), Create(punto), Write(label_punto))
        
        info4 = Text(".animate.rotate(TAU/4, about_point=ORIGIN)", 
                    font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info4))
        
        for _ in range(4):
            self.play(cuadrado.animate.rotate(TAU/4, about_point=ORIGIN))
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateFlip

class AnimateFlip(Scene):
    def construct(self):
        titulo = Text("flip() - Voltear/Reflejar", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objeto asimétrico para ver el flip claramente
        texto = Text("Manim", font_size=48, color=BLUE)
        self.play(Write(texto))
        
        # Flip horizontal (respecto al eje Y - UP)
        info1 = Text(".animate.flip(UP) - Voltear horizontalmente", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(texto.animate.flip(UP))
        self.wait(0.5)
        
        # Flip vertical (respecto al eje X - RIGHT)
        self.play(FadeOut(info1))
        info2 = Text(".animate.flip(RIGHT) - Voltear verticalmente", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(texto.animate.flip(RIGHT))
        self.wait(0.5)
        
        # Restaurar
        self.play(texto.animate.flip(UP))
        self.play(texto.animate.flip(RIGHT))
        
        self.play(FadeOut(texto), FadeOut(info2))
        
        # Demostración con forma asimétrica
        triangulo = Triangle(color=GREEN, fill_opacity=0.7).scale(1.5)
        flecha = Arrow(triangulo.get_center(), triangulo.get_top(), 
                      color=RED, buff=0)
        grupo = VGroup(triangulo, flecha)
        
        self.play(Create(grupo))
        
        info3 = Text("flip() es útil para formas asimétricas", 
                    font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        
        self.play(grupo.animate.flip(RIGHT))
        self.play(grupo.animate.flip(UP))
        
        self.wait()
````

## 5. Métodos de Color con .animate

Los métodos de color permiten cambiar la apariencia visual de los objetos de forma animada.

### Métodos principales:
- **`set_color(color)`**: Cambiar el color general del objeto
- **`set_fill(color, opacity)`**: Cambiar color y opacidad del relleno
- **`set_stroke(color, width, opacity)`**: Cambiar el borde/contorno
- **`set_opacity(valor)`**: Cambiar opacidad general

### Colores disponibles:
Manim incluye constantes como `RED`, `BLUE`, `GREEN`, `YELLOW`, `PURPLE`, `ORANGE`, `PINK`, `WHITE`, `BLACK`, `GREY`, y variantes como `RED_A`, `RED_B`, `RED_C`, `RED_D`, `RED_E`.

````python
%%manim -qm -v WARNING AnimateSetColor

class AnimateSetColor(Scene):
    def construct(self):
        titulo = Text("set_color() - Cambio de Color", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objetos
        circulo = Circle(radius=1, color=BLUE, fill_opacity=0.7)
        cuadrado = Square(side_length=1.5, color=RED, fill_opacity=0.7)
        triangulo = Triangle(color=GREEN, fill_opacity=0.7)
        
        objetos = VGroup(circulo, cuadrado, triangulo).arrange(RIGHT, buff=1.5)
        self.play(*[Create(obj) for obj in objetos])
        
        # Cambio de color básico
        colores = [YELLOW, PURPLE, ORANGE, PINK, TEAL]
        
        for color in colores:
            nombre_color = {
                YELLOW: "YELLOW", PURPLE: "PURPLE", ORANGE: "ORANGE",
                PINK: "PINK", TEAL: "TEAL"
            }[color]
            
            info = Text(f".animate.set_color({nombre_color})", 
                       font_size=18, color=YELLOW).to_edge(DOWN)
            self.play(Write(info))
            self.play(
                circulo.animate.set_color(color),
                cuadrado.animate.set_color(color),
                triangulo.animate.set_color(color)
            )
            self.play(FadeOut(info))
        
        # Colores diferentes para cada objeto
        info2 = Text("Colores diferentes para cada objeto", 
                    font_size=18, color=WHITE).to_edge(DOWN)
        self.play(Write(info2))
        self.play(
            circulo.animate.set_color(RED),
            cuadrado.animate.set_color(BLUE),
            triangulo.animate.set_color(GREEN)
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimateSetFillStroke

class AnimateSetFillStroke(Scene):
    def construct(self):
        titulo = Text("set_fill() y set_stroke()", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear círculo con propiedades visibles
        circulo = Circle(radius=1.5, stroke_width=8, 
                        stroke_color=WHITE, fill_color=BLUE, fill_opacity=0.5)
        self.play(Create(circulo))
        
        # Cambiar fill
        info1 = Text(".animate.set_fill(RED, opacity=0.8)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(circulo.animate.set_fill(RED, opacity=0.8))
        self.wait(0.5)
        
        # Cambiar stroke
        self.play(FadeOut(info1))
        info2 = Text(".animate.set_stroke(YELLOW, width=12)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(circulo.animate.set_stroke(YELLOW, width=12))
        self.wait(0.5)
        
        # Ambos al mismo tiempo
        self.play(FadeOut(info2))
        info3 = Text("set_fill y set_stroke combinados", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(
            circulo.animate.set_fill(GREEN, opacity=0.6).set_stroke(PURPLE, width=6)
        )
        self.wait(0.5)
        
        self.play(FadeOut(circulo), FadeOut(info3))
        
        # Demostración de stroke_opacity
        cuadrado = Square(side_length=2, stroke_width=10, 
                         stroke_color=BLUE, fill_opacity=0)
        self.play(Create(cuadrado))
        
        info4 = Text(".animate.set_stroke(opacity=0.3)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info4))
        self.play(cuadrado.animate.set_stroke(opacity=0.3))
        self.play(cuadrado.animate.set_stroke(opacity=1))
        
        self.wait()
````

## 6. Métodos de Opacidad con .animate

La opacidad permite controlar la transparencia de los objetos, creando efectos de aparición/desaparición gradual.

### Métodos principales:
- **`set_opacity(valor)`**: Opacidad general (0 = invisible, 1 = opaco)
- **`fade(oscuridad)`**: Atenuar el objeto
- **`set_fill(opacity=valor)`**: Opacidad solo del relleno
- **`set_stroke(opacity=valor)`**: Opacidad solo del contorno

````python
%%manim -qm -v WARNING AnimateOpacity

class AnimateOpacity(Scene):
    def construct(self):
        titulo = Text("set_opacity() - Control de Transparencia", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objetos con diferentes colores
        cuadrado = Square(side_length=2, color=BLUE, fill_opacity=1)
        circulo = Circle(radius=1, color=RED, fill_opacity=1)
        circulo.shift(RIGHT * 0.7)  # Superponer parcialmente
        
        grupo = VGroup(cuadrado, circulo)
        self.play(Create(cuadrado), Create(circulo))
        
        # Demostrar opacidad del círculo
        info1 = Text("Cambiar opacidad permite ver superposición", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        
        # Reducir opacidad gradualmente
        for opacidad in [0.8, 0.6, 0.4, 0.2]:
            self.play(circulo.animate.set_opacity(opacidad), run_time=0.5)
        
        # Restaurar
        self.play(circulo.animate.set_opacity(1))
        
        self.play(FadeOut(info1))
        
        # Opacidad de ambos objetos
        info2 = Text(".animate.set_opacity() en múltiples objetos", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        
        self.play(
            cuadrado.animate.set_opacity(0.5),
            circulo.animate.set_opacity(0.5)
        )
        self.wait(0.5)
        
        # Fade hacia transparencia total
        self.play(FadeOut(info2))
        info3 = Text("Fade hacia invisible", font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        
        self.play(grupo.animate.set_opacity(0), run_time=2)
        self.play(grupo.animate.set_opacity(1), run_time=2)
        
        self.wait()
````

## 7. Encadenamiento de Métodos

Una de las características más poderosas de `.animate` es la capacidad de encadenar múltiples métodos en una sola animación.

### Sintaxis:
```python
self.play(objeto.animate.metodo1().metodo2().metodo3())
```

### Ventajas del encadenamiento:
- **Sincronización perfecta**: Todos los cambios ocurren simultáneamente
- **Código más limpio**: Una línea en lugar de múltiples animaciones
- **Transiciones fluidas**: Los cambios se interpolan juntos

### ⚠️ Importante:
El orden de los métodos puede afectar el resultado en algunos casos.

````python
%%manim -qm -v WARNING AnimateChaining

class AnimateChaining(Scene):
    def construct(self):
        titulo = Text("Encadenamiento de Métodos", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear objeto inicial
        cuadrado = Square(side_length=1, color=BLUE, fill_opacity=0.7)
        self.play(Create(cuadrado))
        
        # Ejemplo 1: Mover + Escalar
        info1 = Text(".animate.shift(RIGHT*2).scale(2)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(cuadrado.animate.shift(RIGHT * 2).scale(2))
        self.wait(0.5)
        
        # Ejemplo 2: Rotar + Cambiar color
        self.play(FadeOut(info1))
        info2 = Text(".animate.rotate(PI/4).set_color(RED)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(cuadrado.animate.rotate(PI/4).set_color(RED))
        self.wait(0.5)
        
        # Ejemplo 3: Múltiples cambios
        self.play(FadeOut(info2))
        info3 = Text(".animate.move_to(LEFT*2).scale(0.5).set_color(GREEN).rotate(-PI/4)", 
                    font_size=12, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(
            cuadrado.animate.move_to(LEFT * 2).scale(0.5).set_color(GREEN).rotate(-PI/4)
        )
        self.wait(0.5)
        
        self.play(FadeOut(cuadrado), FadeOut(info3))
        
        # Comparación: separado vs encadenado
        circulo1 = Circle(radius=0.5, color=BLUE, fill_opacity=0.7).shift(LEFT * 3)
        circulo2 = Circle(radius=0.5, color=RED, fill_opacity=0.7).shift(RIGHT * 3)
        
        label1 = Text("Separado", font_size=14).next_to(circulo1, UP)
        label2 = Text("Encadenado", font_size=14).next_to(circulo2, UP)
        
        self.play(Create(circulo1), Create(circulo2), Write(label1), Write(label2))
        
        # Animaciones separadas vs encadenadas
        info4 = Text("Mismo resultado, diferente ejecución", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info4))
        
        # Separado (secuencial)
        self.play(circulo1.animate.shift(UP))
        self.play(circulo1.animate.scale(1.5))
        self.play(circulo1.animate.set_color(YELLOW))
        
        # Encadenado (simultáneo)
        self.play(circulo2.animate.shift(UP).scale(1.5).set_color(YELLOW))
        
        self.wait()
````

## 8. Animaciones Simultáneas con .animate

Puedes animar múltiples objetos al mismo tiempo pasando varias animaciones `.animate` a `self.play()`.

### Sintaxis:
```python
self.play(
    objeto1.animate.metodo(),
    objeto2.animate.metodo(),
    objeto3.animate.metodo()
)
```

### Casos de uso:
- Sincronizar movimientos de múltiples objetos
- Crear efectos de grupo coordinados
- Transiciones complejas de escena

````python
%%manim -qm -v WARNING AnimateSimultaneous

class AnimateSimultaneous(Scene):
    def construct(self):
        titulo = Text("Animaciones Simultáneas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear múltiples objetos
        circulo = Circle(radius=0.5, color=RED, fill_opacity=0.7).shift(LEFT * 4)
        cuadrado = Square(side_length=1, color=BLUE, fill_opacity=0.7)
        triangulo = Triangle(color=GREEN, fill_opacity=0.7).scale(0.7).shift(RIGHT * 4)
        
        self.play(Create(circulo), Create(cuadrado), Create(triangulo))
        
        # Movimientos simultáneos
        info1 = Text("Tres objetos moviéndose al mismo tiempo", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        
        self.play(
            circulo.animate.shift(UP * 2),
            cuadrado.animate.shift(DOWN * 2),
            triangulo.animate.shift(UP * 2)
        )
        
        # Cambios múltiples simultáneos
        self.play(FadeOut(info1))
        info2 = Text("Cambios de posición, escala y color simultáneos", 
                    font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        
        self.play(
            circulo.animate.shift(DOWN * 2).scale(2).set_color(YELLOW),
            cuadrado.animate.shift(UP * 2).rotate(PI/4).set_color(PURPLE),
            triangulo.animate.shift(DOWN * 2).scale(1.5).set_color(ORANGE)
        )
        
        # Efecto de convergencia
        self.play(FadeOut(info2))
        info3 = Text("Convergencia al centro", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        
        self.play(
            circulo.animate.move_to(ORIGIN + LEFT),
            cuadrado.animate.move_to(ORIGIN),
            triangulo.animate.move_to(ORIGIN + RIGHT)
        )
        
        # Dispersión
        self.play(
            circulo.animate.to_corner(UL).scale(0.5),
            cuadrado.animate.to_corner(DR).scale(0.5),
            triangulo.animate.to_corner(UR).scale(0.5)
        )
        
        self.wait()
````

## 9. Limitaciones de .animate

Aunque `.animate` es muy poderoso, tiene algunas limitaciones importantes que debes conocer.

### ❌ Lo que .animate NO puede hacer:

1. **Métodos que no modifican el estado**:
   - `get_center()`, `get_width()` - Solo retornan información
   - `copy()` - Crea un nuevo objeto

2. **Creación/Destrucción de objetos**:
   - No puede crear objetos (`Create`, `Write`)
   - No puede eliminar objetos (`Uncreate`, `FadeOut`)

3. **Cambios estructurales**:
   - Añadir/quitar submobjects
   - Cambiar el número de puntos de un path

4. **Interpolación de algunos tipos**:
   - Cambio de texto (usar `TransformMatchingTex`)
   - Cambio entre tipos de mobject diferentes

### ⚠️ Problemas comunes:
- `.animate` interpola entre el estado actual y el final
- A veces la interpolación no es la esperada (especialmente con paths)

````python
%%manim -qm -v WARNING AnimateLimitations

class AnimateLimitations(Scene):
    def construct(self):
        titulo = Text("Limitaciones de .animate", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Limitación 1: No funciona para creación
        nota1 = Text("❌ .animate NO sirve para crear objetos", 
                    font_size=18, color=RED).shift(UP)
        self.play(Write(nota1))
        
        circulo = Circle(radius=1, color=BLUE, fill_opacity=0.7)
        # Esto NO funcionaría: self.play(circulo.animate)  # Error!
        # Correcto:
        self.play(Create(circulo))
        
        correcto1 = Text("✓ Usar Create(), Write(), FadeIn()", 
                        font_size=16, color=GREEN).next_to(nota1, DOWN)
        self.play(Write(correcto1))
        self.wait()
        
        self.play(FadeOut(nota1), FadeOut(correcto1))
        
        # Limitación 2: Interpolación inesperada
        nota2 = Text("⚠️ La interpolación puede ser inesperada", 
                    font_size=18, color=YELLOW).shift(UP * 2)
        self.play(Write(nota2))
        
        # Ejemplo: rotate puede tomar camino corto o largo
        flecha = Arrow(ORIGIN, RIGHT * 2, color=GREEN, buff=0)
        flecha.shift(DOWN)
        self.play(Create(flecha))
        
        info = Text("rotate(PI*1.9) - Casi 360°, pero toma camino corto", 
                   font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        # La animación puede no hacer el giro completo esperado
        self.play(flecha.animate.rotate(PI * 1.9), run_time=2)
        
        self.play(FadeOut(flecha), FadeOut(info), FadeOut(nota2), FadeOut(circulo))
        
        # Limitación 3: No cambia estructuras
        nota3 = Text("❌ No puede cambiar la estructura de un objeto", 
                    font_size=18, color=RED).shift(UP)
        self.play(Write(nota3))
        
        # Ejemplo con VGroup
        grupo = VGroup(
            Circle(radius=0.3, color=RED),
            Square(side_length=0.5, color=BLUE),
            Triangle(color=GREEN).scale(0.4)
        ).arrange(RIGHT, buff=0.5)
        
        self.play(Create(grupo))
        
        correcto3 = Text("✓ Usar Transform o AnimationGroup para cambios estructurales", 
                        font_size=14, color=GREEN).to_edge(DOWN)
        self.play(Write(correcto3))
        
        # Esto SÍ funciona: cambiar propiedades del grupo
        self.play(grupo.animate.arrange(DOWN, buff=0.3))
        
        self.wait()
````

## 10. .animate vs Transform

Es importante entender cuándo usar `.animate` y cuándo usar `Transform`.

### Cuándo usar .animate:
- ✅ Cambios de posición, escala, rotación
- ✅ Cambios de color y opacidad
- ✅ Encadenar múltiples cambios simples
- ✅ Mantener la identidad del objeto

### Cuándo usar Transform:
- ✅ Convertir un objeto en otro diferente
- ✅ Cambios estructurales complejos
- ✅ Transformaciones de texto
- ✅ Cambio del número de elementos

### Comparación:
| Característica | .animate | Transform |
|---------------|----------|-----------|
| Sintaxis | `obj.animate.metodo()` | `Transform(obj1, obj2)` |
| Objeto resultante | El mismo objeto | Segundo objeto |
| Interpolación | Automática | Punto a punto |
| Flexibilidad | Media | Alta |

````python
%%manim -qm -v WARNING AnimateVsTransform

class AnimateVsTransform(Scene):
    def construct(self):
        titulo = Text(".animate vs Transform", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear dos círculos para comparación
        circulo1 = Circle(radius=0.8, color=BLUE, fill_opacity=0.7).shift(LEFT * 3)
        circulo2 = Circle(radius=0.8, color=BLUE, fill_opacity=0.7).shift(RIGHT * 3)
        
        label1 = Text(".animate", font_size=16).next_to(circulo1, UP)
        label2 = Text("Transform", font_size=16).next_to(circulo2, UP)
        
        self.play(Create(circulo1), Create(circulo2), Write(label1), Write(label2))
        
        # Caso 1: Cambio de color y escala
        info1 = Text("Caso 1: Cambio de color y escala", 
                    font_size=18, color=YELLOW).shift(DOWN * 2)
        self.play(Write(info1))
        
        # Con .animate
        self.play(circulo1.animate.scale(1.5).set_color(RED))
        
        # Con Transform
        circulo2_nuevo = Circle(radius=1.2, color=RED, fill_opacity=0.7).shift(RIGHT * 3)
        self.play(Transform(circulo2, circulo2_nuevo))
        
        resultado1 = Text("Ambos logran el mismo resultado", 
                         font_size=14, color=GREEN).shift(DOWN * 3)
        self.play(Write(resultado1))
        self.wait()
        
        self.play(FadeOut(info1), FadeOut(resultado1))
        
        # Caso 2: Cambiar forma (aquí Transform es necesario)
        info2 = Text("Caso 2: Cambiar la forma del objeto", 
                    font_size=18, color=YELLOW).shift(DOWN * 2)
        self.play(Write(info2))
        
        # .animate NO puede cambiar un círculo en cuadrado
        # (podemos simular con escala, pero no es lo mismo)
        self.play(circulo1.animate.stretch(1.2, dim=0).stretch(0.8, dim=1))
        
        # Transform SÍ puede
        cuadrado = Square(side_length=1.5, color=RED, fill_opacity=0.7).shift(RIGHT * 3)
        self.play(Transform(circulo2, cuadrado))
        
        resultado2 = Text("Transform convierte círculo en cuadrado realmente", 
                         font_size=14, color=GREEN).shift(DOWN * 3)
        self.play(Write(resultado2))
        self.wait()
        
        self.play(FadeOut(circulo1), FadeOut(circulo2), FadeOut(label1), 
                 FadeOut(label2), FadeOut(info2), FadeOut(resultado2))
        
        # Resumen visual
        resumen = VGroup(
            Text(".animate:", font_size=20, color=BLUE),
            Text("• Mismo objeto modificado", font_size=16),
            Text("• Sintaxis simple", font_size=16),
            Text("• Solo cambios de propiedades", font_size=16),
        ).arrange(DOWN, aligned_edge=LEFT).shift(LEFT * 3)
        
        resumen2 = VGroup(
            Text("Transform:", font_size=20, color=RED),
            Text("• Puede cambiar a otro objeto", font_size=16),
            Text("• Más flexible", font_size=16),
            Text("• Interpola punto a punto", font_size=16),
        ).arrange(DOWN, aligned_edge=LEFT).shift(RIGHT * 3)
        
        self.play(Write(resumen), Write(resumen2))
        
        self.wait()
````

## 11. Parámetros de Animación con .animate

`.animate` acepta los mismos parámetros que cualquier animación en `self.play()`.

### Parámetros comunes:
- **`run_time`**: Duración de la animación
- **`rate_func`**: Función de interpolación (easing)
- **`lag_ratio`**: Retraso entre sub-animaciones

### Rate Functions populares:
- `linear` - Velocidad constante
- `smooth` - Suave (por defecto)
- `rush_into` - Acelera al final
- `rush_from` - Desacelera al final  
- `there_and_back` - Va y vuelve
- `wiggle` - Efecto de rebote

````python
%%manim -qm -v WARNING AnimateParameters

class AnimateParameters(Scene):
    def construct(self):
        titulo = Text("Parámetros de Animación", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # run_time
        cuadrado = Square(side_length=1.5, color=BLUE, fill_opacity=0.7)
        cuadrado.shift(LEFT * 4)
        self.play(Create(cuadrado))
        
        info1 = Text("run_time=3 (3 segundos)", font_size=18, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(cuadrado.animate.shift(RIGHT * 8), run_time=3)
        
        self.play(FadeOut(info1))
        
        # rate_func - diferentes funciones
        self.play(cuadrado.animate.move_to(LEFT * 4))
        
        rate_funcs = [
            (linear, "linear"),
            (smooth, "smooth"),
            (rush_into, "rush_into"),
            (rush_from, "rush_from"),
        ]
        
        for func, nombre in rate_funcs:
            info = Text(f"rate_func={nombre}", font_size=18, color=YELLOW).to_edge(DOWN)
            self.play(Write(info))
            self.play(cuadrado.animate.shift(RIGHT * 8), rate_func=func, run_time=1.5)
            self.play(FadeOut(info))
            self.play(cuadrado.animate.move_to(LEFT * 4), run_time=0.3)
        
        self.play(FadeOut(cuadrado))
        
        # there_and_back - efecto especial
        circulo = Circle(radius=1, color=RED, fill_opacity=0.7)
        self.play(Create(circulo))
        
        info2 = Text("rate_func=there_and_back (va y vuelve)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(circulo.animate.scale(2).set_color(YELLOW), 
                 rate_func=there_and_back, run_time=2)
        
        self.play(FadeOut(info2))
        
        # wiggle
        info3 = Text("rate_func=wiggle (rebote)", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(circulo.animate.shift(RIGHT * 3), rate_func=wiggle, run_time=2)
        
        self.wait()
````

## 12. Ejercicios Prácticos

Ahora que conoces `.animate`, practica con estos ejercicios progresivos.

````python
%%manim -qm -v WARNING Ejercicio1_Cuadrado_Bailarin

# Ejercicio 1: Cuadrado Bailarín
# Crear un cuadrado que:
# 1. Se mueva a la derecha
# 2. Rote 45 grados
# 3. Cambie a color rojo
# 4. Se escale al doble
# 5. Vuelva al centro

class Ejercicio1_Cuadrado_Bailarin(Scene):
    def construct(self):
        titulo = Text("Ejercicio 1: Cuadrado Bailarín", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        cuadrado = Square(side_length=1.5, color=BLUE, fill_opacity=0.7)
        self.play(Create(cuadrado))
        
        # Paso 1: Mover a la derecha
        self.play(cuadrado.animate.shift(RIGHT * 3))
        
        # Paso 2: Rotar 45 grados
        self.play(cuadrado.animate.rotate(PI/4))
        
        # Paso 3: Cambiar a rojo
        self.play(cuadrado.animate.set_color(RED))
        
        # Paso 4: Escalar al doble
        self.play(cuadrado.animate.scale(2))
        
        # Paso 5: Volver al centro (todo junto)
        self.play(cuadrado.animate.move_to(ORIGIN).scale(0.5).rotate(-PI/4).set_color(BLUE))
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio2_Orbita_Planetaria

# Ejercicio 2: Órbita Planetaria
# Simular un planeta orbitando alrededor de un sol usando .animate

class Ejercicio2_Orbita_Planetaria(Scene):
    def construct(self):
        titulo = Text("Ejercicio 2: Órbita Planetaria", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        # Sol en el centro
        sol = Circle(radius=0.5, color=YELLOW, fill_opacity=1)
        sol_glow = Circle(radius=0.7, color=ORANGE, fill_opacity=0.3)
        sol_grupo = VGroup(sol_glow, sol)
        
        # Planeta
        planeta = Circle(radius=0.2, color=BLUE, fill_opacity=1)
        planeta.shift(RIGHT * 2.5)
        
        # Órbita (referencia visual)
        orbita = Circle(radius=2.5, color=WHITE, stroke_opacity=0.3)
        
        self.play(Create(orbita), Create(sol_grupo), Create(planeta))
        
        # Animar órbita usando rotación alrededor del sol
        for _ in range(2):  # 2 órbitas completas
            self.play(
                planeta.animate.rotate(PI/2, about_point=ORIGIN),
                run_time=1
            )
        
        # Hacer que el planeta se acerque al sol
        self.play(planeta.animate.scale(1.5).move_to(RIGHT * 1))
        
        # Y luego se aleje
        self.play(planeta.animate.scale(0.5).move_to(RIGHT * 3.5))
        
        # Última órbita
        self.play(
            planeta.animate.rotate(TAU, about_point=ORIGIN),
            run_time=3
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio3_Transformacion_Coordinada

# Ejercicio 3: Transformación Coordinada
# Múltiples objetos moviéndose de forma coordinada

class Ejercicio3_Transformacion_Coordinada(Scene):
    def construct(self):
        titulo = Text("Ejercicio 3: Transformación Coordinada", font_size=26).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear 5 círculos en línea
        circulos = VGroup(*[
            Circle(radius=0.4, color=color, fill_opacity=0.8)
            for color in [RED, ORANGE, YELLOW, GREEN, BLUE]
        ]).arrange(RIGHT, buff=0.5)
        
        self.play(*[Create(c) for c in circulos])
        
        # Movimiento de onda
        info1 = Text("Efecto de onda hacia arriba", font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info1))
        
        for i, circulo in enumerate(circulos):
            self.play(
                circulo.animate.shift(UP * 1.5),
                run_time=0.3
            )
        
        self.play(FadeOut(info1))
        
        # Todos vuelven al mismo tiempo
        info2 = Text("Todos vuelven simultáneamente", font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info2))
        self.play(*[c.animate.shift(DOWN * 1.5) for c in circulos])
        
        self.play(FadeOut(info2))
        
        # Escala en cascada
        info3 = Text("Escala en cascada", font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info3))
        
        self.play(
            circulos[0].animate.scale(1.5),
            circulos[1].animate.scale(1.3),
            circulos[2].animate.scale(1.5),
            circulos[3].animate.scale(1.3),
            circulos[4].animate.scale(1.5)
        )
        
        # Reorganizar en círculo
        self.play(FadeOut(info3))
        info4 = Text("Reorganizar en formación circular", font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info4))
        
        angulos = [i * TAU / 5 for i in range(5)]
        radio = 2
        
        self.play(*[
            circulos[i].animate.move_to(
                np.array([radio * np.cos(angulos[i]), radio * np.sin(angulos[i]), 0])
            ).scale(1/1.5 if i % 2 == 0 else 1/1.3)
            for i in range(5)
        ])
        
        # Rotación del grupo
        self.play(circulos.animate.rotate(TAU/5, about_point=ORIGIN), run_time=2)
        
        self.wait()
````

## 13. Referencia Rápida

### Sintaxis básica:
```python
self.play(objeto.animate.metodo())
```

### Métodos de posición:
```python
.animate.shift(vector)           # Desplazamiento relativo
.animate.move_to(posición)       # Posición absoluta
.animate.next_to(obj, dirección) # Relativo a otro objeto
.animate.to_edge(dirección)      # Ir al borde
.animate.to_corner(esquina)      # Ir a esquina
```

### Métodos de escala:
```python
.animate.scale(factor)           # Escalar uniformemente
.animate.stretch(factor, dim)    # Estirar en una dimensión
.animate.scale_to_fit_width(w)   # Ajustar al ancho
.animate.scale_to_fit_height(h)  # Ajustar a la altura
```

### Métodos de rotación:
```python
.animate.rotate(ángulo)          # Rotar en radianes
.animate.rotate(ángulo, about_point=punto)  # Rotar alrededor de punto
.animate.flip(eje)               # Voltear
```

### Métodos de color:
```python
.animate.set_color(color)        # Cambiar color
.animate.set_fill(color, opacity) # Cambiar relleno
.animate.set_stroke(color, width) # Cambiar contorno
.animate.set_opacity(valor)      # Cambiar opacidad
```

### Encadenamiento:
```python
.animate.shift(UP).scale(2).set_color(RED).rotate(PI/4)
```

### Parámetros de animación:
```python
self.play(obj.animate.metodo(), run_time=2, rate_func=smooth)
```
