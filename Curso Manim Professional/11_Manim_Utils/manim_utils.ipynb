## 1. Rate Functions - Funciones de Interpolación

Las **Rate Functions** controlan cómo progresa una animación en el tiempo. En lugar de movimiento lineal, puedes crear efectos de aceleración, desaceleración, rebotes, etc.

### Rate Functions incorporadas:
- `linear` - Velocidad constante
- `smooth` - Suave (por defecto)
- `rush_into` - Acelera hacia el final
- `rush_from` - Desacelera desde el inicio
- `slow_into` - Desacelera hacia el final
- `double_smooth` - Doble suavizado
- `there_and_back` - Va y vuelve
- `there_and_back_with_pause` - Va, pausa, vuelve
- `running_start` - Arranca con impulso
- `wiggle` - Efecto de temblor
- `lingering` - Se queda al final
- `exponential_decay` - Decaimiento exponencial

````python
%%manim -qm -v WARNING RateFunctionsDemo

class RateFunctionsDemo(Scene):
    def construct(self):
        titulo = Text("Rate Functions - Comparación", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear varios objetos para comparar
        rate_funcs = [
            (linear, "linear", RED),
            (smooth, "smooth", BLUE),
            (rush_into, "rush_into", GREEN),
            (rush_from, "rush_from", YELLOW),
            (there_and_back, "there_and_back", PURPLE),
        ]
        
        # Crear círculos y labels
        circulos = VGroup()
        labels = VGroup()
        
        for i, (func, nombre, color) in enumerate(rate_funcs):
            y_pos = 2 - i * 1
            circulo = Circle(radius=0.3, color=color, fill_opacity=0.8)
            circulo.move_to(LEFT * 5 + UP * y_pos)
            label = Text(nombre, font_size=14).next_to(circulo, LEFT)
            circulos.add(circulo)
            labels.add(label)
        
        self.play(*[Create(c) for c in circulos], *[Write(l) for l in labels])
        
        # Animar todos con diferentes rate functions
        animaciones = [
            circulos[i].animate.shift(RIGHT * 10)
            for i in range(len(rate_funcs))
        ]
        
        # Aplicar rate functions
        self.play(
            animaciones[0].set_rate_func(linear),
            animaciones[1].set_rate_func(smooth),
            animaciones[2].set_rate_func(rush_into),
            animaciones[3].set_rate_func(rush_from),
            animaciones[4].set_rate_func(there_and_back),
            run_time=3
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING CustomRateFunction

class CustomRateFunction(Scene):
    def construct(self):
        titulo = Text("Rate Functions Personalizadas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear una rate function personalizada
        def bounce(t):
            """Efecto de rebote personalizado"""
            if t < 0.5:
                return 2 * t * t
            else:
                return 1 - 2 * (1 - t) ** 2
        
        def elastic(t):
            """Efecto elástico"""
            return 1 - np.cos(t * 2 * PI) * (1 - t)
        
        cuadrado = Square(side_length=1.5, color=GREEN, fill_opacity=0.7)
        cuadrado.shift(LEFT * 4)
        self.play(Create(cuadrado))
        
        # Usar rate function personalizada
        info1 = Text("Rate Function personalizada: bounce", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(cuadrado.animate.shift(RIGHT * 8), rate_func=bounce, run_time=2)
        
        self.play(FadeOut(info1))
        self.play(cuadrado.animate.move_to(LEFT * 4), run_time=0.3)
        
        info2 = Text("Rate Function personalizada: elastic", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(cuadrado.animate.shift(RIGHT * 8), rate_func=elastic, run_time=2)
        
        # Mostrar código
        codigo = Code(
            code='''def bounce(t):
    if t < 0.5:
        return 2 * t * t
    else:
        return 1 - 2 * (1-t)**2''',
            language="python",
            font_size=14,
            background="window"
        ).shift(DOWN)
        
        self.play(FadeOut(cuadrado), FadeOut(info2))
        self.play(Create(codigo))
        
        self.wait()
````

## 2. Color Utilities - Manipulación de Colores

Manim proporciona herramientas poderosas para trabajar con colores.

### Funciones de color:
- `color_to_rgb(color)` - Convierte color a RGB
- `rgb_to_color(rgb)` - Convierte RGB a color de Manim
- `color_to_rgba(color)` - Incluye canal alpha
- `interpolate_color(c1, c2, alpha)` - Interpola entre dos colores
- `invert_color(color)` - Invierte el color
- `color_gradient(colors, n)` - Genera gradiente de n colores

### Colores predefinidos:
Manim tiene muchos colores como constantes: `RED`, `BLUE`, `GREEN`, `YELLOW`, `PURPLE`, `ORANGE`, `PINK`, `TEAL`, `GOLD`, `MAROON`, etc.

Cada color tiene variantes de A (más claro) a E (más oscuro): `RED_A`, `RED_B`, `RED_C`, `RED_D`, `RED_E`

````python
%%manim -qm -v WARNING ColorUtilitiesDemo

class ColorUtilitiesDemo(Scene):
    def construct(self):
        titulo = Text("Color Utilities", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Demostrar variantes de color
        subtitulo1 = Text("Variantes de RED (A→E)", font_size=20).shift(UP * 2)
        self.play(Write(subtitulo1))
        
        colores_red = [RED_A, RED_B, RED_C, RED_D, RED_E]
        nombres_red = ["RED_A", "RED_B", "RED_C", "RED_D", "RED_E"]
        
        cuadrados_red = VGroup()
        for i, (color, nombre) in enumerate(zip(colores_red, nombres_red)):
            cuadrado = Square(side_length=1, color=color, fill_opacity=1)
            label = Text(nombre, font_size=10).next_to(cuadrado, DOWN, buff=0.1)
            grupo = VGroup(cuadrado, label)
            cuadrados_red.add(grupo)
        
        cuadrados_red.arrange(RIGHT, buff=0.3).shift(UP * 0.5)
        self.play(*[Create(c) for c in cuadrados_red])
        self.wait()
        
        self.play(FadeOut(cuadrados_red), FadeOut(subtitulo1))
        
        # Interpolación de colores
        subtitulo2 = Text("Interpolación de colores", font_size=20).shift(UP * 2)
        self.play(Write(subtitulo2))
        
        # Crear gradiente visual
        n_colores = 10
        cuadrados_grad = VGroup()
        
        for i in range(n_colores):
            alpha = i / (n_colores - 1)
            color = interpolate_color(BLUE, RED, alpha)
            cuadrado = Square(side_length=0.8, fill_color=color, fill_opacity=1, stroke_width=0)
            cuadrados_grad.add(cuadrado)
        
        cuadrados_grad.arrange(RIGHT, buff=0.1)
        self.play(Create(cuadrados_grad))
        
        info = Text("interpolate_color(BLUE, RED, alpha)", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

````python
%%manim -qm -v WARNING ColorGradientDemo

class ColorGradientDemo(Scene):
    def construct(self):
        titulo = Text("Color Gradient", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Gradiente con múltiples colores
        colores_base = [RED, ORANGE, YELLOW, GREEN, BLUE, PURPLE]
        
        # Crear un arcoíris de círculos
        circulos = VGroup()
        n = 30
        
        for i in range(n):
            # Calcular qué segmento del gradiente usar
            pos_relativa = i / (n - 1) * (len(colores_base) - 1)
            idx = int(pos_relativa)
            alpha = pos_relativa - idx
            
            if idx >= len(colores_base) - 1:
                color = colores_base[-1]
            else:
                color = interpolate_color(colores_base[idx], colores_base[idx + 1], alpha)
            
            circulo = Circle(radius=0.3, fill_color=color, fill_opacity=1, stroke_width=0)
            circulos.add(circulo)
        
        circulos.arrange_in_grid(rows=3, buff=0.2)
        self.play(LaggedStart(*[GrowFromCenter(c) for c in circulos], lag_ratio=0.05))
        
        self.wait()
        
        # Animar cambio de colores
        info = Text("Animación de gradiente", font_size=18, color=WHITE).to_edge(DOWN)
        self.play(Write(info))
        
        # Rotar los colores
        for _ in range(2):
            nuevos_colores = colores_base[1:] + [colores_base[0]]
            animaciones = []
            
            for i, circulo in enumerate(circulos):
                pos_relativa = i / (n - 1) * (len(nuevos_colores) - 1)
                idx = int(pos_relativa)
                alpha = pos_relativa - idx
                
                if idx >= len(nuevos_colores) - 1:
                    color = nuevos_colores[-1]
                else:
                    color = interpolate_color(nuevos_colores[idx], nuevos_colores[idx + 1], alpha)
                
                animaciones.append(circulo.animate.set_fill(color))
            
            self.play(*animaciones, run_time=1)
            colores_base = nuevos_colores
        
        self.wait()
````

## 3. Bezier Curves - Curvas de Bézier

Las curvas de Bézier son fundamentales en Manim para crear paths suaves y personalizados.

### Funciones disponibles:
- `bezier(points)` - Crea una función de curva de Bézier
- `partial_bezier_points(points, a, b)` - Obtiene puntos de una porción de la curva
- `interpolate(start, end, alpha)` - Interpola entre dos valores

### Tipos de curvas:
- **Lineal (2 puntos)**: Línea recta
- **Cuadrática (3 puntos)**: Una curva con un punto de control
- **Cúbica (4 puntos)**: Curva con dos puntos de control (más común)

````python
%%manim -qm -v WARNING BezierCurvesDemo

class BezierCurvesDemo(Scene):
    def construct(self):
        titulo = Text("Curvas de Bézier", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Puntos de control para curva cúbica
        p0 = np.array([-4, -2, 0])
        p1 = np.array([-2, 2, 0])
        p2 = np.array([2, 2, 0])
        p3 = np.array([4, -2, 0])
        
        puntos = [p0, p1, p2, p3]
        
        # Mostrar puntos de control
        dots = VGroup()
        labels = VGroup()
        for i, p in enumerate(puntos):
            dot = Dot(p, color=RED if i in [0, 3] else YELLOW)
            label = Text(f"P{i}", font_size=14).next_to(dot, UP, buff=0.1)
            dots.add(dot)
            labels.add(label)
        
        # Líneas de control
        lineas_control = VGroup(
            Line(p0, p1, color=GREY, stroke_opacity=0.5),
            Line(p1, p2, color=GREY, stroke_opacity=0.5),
            Line(p2, p3, color=GREY, stroke_opacity=0.5)
        )
        
        self.play(Create(dots), Write(labels), Create(lineas_control))
        
        # Crear la curva de Bézier usando CubicBezier
        curva = CubicBezier(p0, p1, p2, p3, color=BLUE)
        
        info = Text("CubicBezier(P0, P1, P2, P3)", font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        self.play(Create(curva), run_time=2)
        
        # Animar un punto recorriendo la curva
        punto_movil = Dot(p0, color=GREEN)
        self.play(Create(punto_movil))
        
        # Función bezier para obtener posición
        bezier_func = bezier(puntos)
        
        self.play(FadeOut(info))
        info2 = Text("Punto recorriendo la curva", font_size=16, color=GREEN).to_edge(DOWN)
        self.play(Write(info2))
        
        # Animar el punto
        self.play(
            MoveAlongPath(punto_movil, curva),
            run_time=3
        )
        
        self.wait()
````

## 4. Space Operations - Operaciones Espaciales

Manim incluye utilidades para trabajar con coordenadas y espacios.

### Constantes de dirección:
- `UP`, `DOWN`, `LEFT`, `RIGHT` - Direcciones cardinales
- `UL`, `UR`, `DL`, `DR` - Esquinas (UpperLeft, UpperRight, etc.)
- `ORIGIN` - Centro de la escena (0, 0, 0)
- `IN`, `OUT` - Profundidad (eje Z)

### Funciones espaciales:
- `normalize(vector)` - Normaliza un vector
- `get_norm(vector)` - Obtiene la longitud del vector
- `angle_of_vector(vector)` - Ángulo del vector en 2D
- `rotate_vector(vector, angle)` - Rota un vector
- `complex_to_R3(z)` - Convierte número complejo a coordenada 3D

````python
%%manim -qm -v WARNING SpaceOperationsDemo

class SpaceOperationsDemo(Scene):
    def construct(self):
        titulo = Text("Operaciones Espaciales", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Mostrar las direcciones constantes
        origen = Dot(ORIGIN, color=WHITE)
        label_origen = Text("ORIGIN", font_size=12).next_to(origen, DOWN, buff=0.1)
        
        # Crear flechas para cada dirección
        direcciones = [
            (UP, "UP", BLUE),
            (DOWN, "DOWN", BLUE),
            (LEFT, "LEFT", RED),
            (RIGHT, "RIGHT", RED),
            (UL, "UL", GREEN),
            (UR, "UR", GREEN),
            (DL, "DL", YELLOW),
            (DR, "DR", YELLOW)
        ]
        
        flechas = VGroup()
        for direccion, nombre, color in direcciones:
            flecha = Arrow(ORIGIN, direccion * 2, color=color, buff=0)
            label = Text(nombre, font_size=12).move_to(direccion * 2.3)
            flechas.add(VGroup(flecha, label))
        
        self.play(Create(origen), Write(label_origen))
        self.play(LaggedStart(*[Create(f) for f in flechas], lag_ratio=0.1))
        
        self.wait()
        self.play(FadeOut(flechas), FadeOut(origen), FadeOut(label_origen))
        
        # Demostrar rotate_vector
        subtitulo = Text("rotate_vector()", font_size=24).shift(UP * 2)
        self.play(Write(subtitulo))
        
        vector_original = Arrow(ORIGIN, RIGHT * 3, color=BLUE, buff=0)
        self.play(Create(vector_original))
        
        # Rotar el vector
        info = Text("Rotando vector 45° cada vez", font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        for _ in range(8):
            self.play(Rotate(vector_original, PI/4, about_point=ORIGIN), run_time=0.5)
        
        self.wait()
````

## 5. Configuración de Manim (Config)

Manim permite configurar muchos aspectos del renderizado a través del objeto `config`.

### Propiedades principales:
- `config.pixel_width` / `config.pixel_height` - Resolución del video
- `config.frame_rate` - Frames por segundo
- `config.background_color` - Color de fondo
- `config.quality` - Calidad de renderizado

### Calidades predefinidas:
- `low_quality` (-ql): 480p, 15fps
- `medium_quality` (-qm): 720p, 30fps
- `high_quality` (-qh): 1080p, 60fps
- `production_quality` (-qp): 1440p, 60fps
- `fourk_quality` (-qk): 4K, 60fps

````python
%%manim -qm -v WARNING ConfigDemo

class ConfigDemo(Scene):
    def construct(self):
        titulo = Text("Información de Configuración", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Mostrar información de configuración actual
        info_config = VGroup(
            Text(f"Resolución: {config.pixel_width} x {config.pixel_height}", font_size=20),
            Text(f"Frame Rate: {config.frame_rate} fps", font_size=20),
            Text(f"Frame Width: {config.frame_width}", font_size=20),
            Text(f"Frame Height: {config.frame_height}", font_size=20),
        ).arrange(DOWN, aligned_edge=LEFT, buff=0.3)
        
        self.play(Write(info_config))
        self.wait()
        
        # Demostrar el tamaño del frame
        self.play(info_config.animate.shift(UP * 1.5))
        
        # Dibujar el borde del frame
        borde = Rectangle(
            width=config.frame_width,
            height=config.frame_height,
            color=YELLOW,
            stroke_width=4
        )
        
        info_frame = Text("Límites del frame visible", 
                         font_size=16, color=YELLOW).to_edge(DOWN)
        
        self.play(Create(borde), Write(info_frame))
        
        # Mostrar las coordenadas de las esquinas
        esquinas_info = VGroup(
            Text(f"(-{config.frame_width/2:.1f}, {config.frame_height/2:.1f})", font_size=12).to_corner(UL),
            Text(f"({config.frame_width/2:.1f}, {config.frame_height/2:.1f})", font_size=12).to_corner(UR),
            Text(f"(-{config.frame_width/2:.1f}, -{config.frame_height/2:.1f})", font_size=12).to_corner(DL),
            Text(f"({config.frame_width/2:.1f}, -{config.frame_height/2:.1f})", font_size=12).to_corner(DR),
        )
        
        self.play(*[Write(e) for e in esquinas_info])
        
        self.wait()
````

## 6. Debug Tools - Herramientas de Depuración

Manim proporciona herramientas útiles para depurar y entender mejor tus escenas.

### Herramientas de debug:
- `NumberPlane()` - Grid de coordenadas
- `Dot()` - Puntos para marcar posiciones
- `index_labels(mobject)` - Muestra índices de submobjects
- `DebugPoints` - Muestra puntos de un path

### Métodos útiles de mobjects:
- `mobject.get_center()` - Obtiene el centro
- `mobject.get_corner(direction)` - Obtiene una esquina
- `mobject.get_edge_center(direction)` - Centro de un borde
- `mobject.get_start()` / `get_end()` - Inicio/fin de líneas

````python
%%manim -qm -v WARNING DebugToolsDemo

class DebugToolsDemo(Scene):
    def construct(self):
        titulo = Text("Herramientas de Debug", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # NumberPlane como grid de referencia
        plano = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-4, 4, 1],
            background_line_style={"stroke_opacity": 0.4}
        )
        self.play(Create(plano, run_time=1))
        
        info1 = Text("NumberPlane - Grid de coordenadas", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.wait()
        
        # Crear un objeto para analizar
        cuadrado = Square(side_length=2, color=BLUE, fill_opacity=0.5)
        cuadrado.shift(RIGHT * 2 + UP)
        self.play(Create(cuadrado), FadeOut(info1))
        
        # Mostrar puntos importantes del objeto
        info2 = Text("Métodos get_center, get_corner, get_edge_center", 
                    font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        
        # Centro
        centro = Dot(cuadrado.get_center(), color=RED)
        label_centro = Text("center", font_size=10, color=RED).next_to(centro, DOWN, buff=0.1)
        
        # Esquinas
        esquinas = VGroup()
        for direction, name in [(UL, "UL"), (UR, "UR"), (DL, "DL"), (DR, "DR")]:
            punto = Dot(cuadrado.get_corner(direction), color=GREEN)
            label = Text(name, font_size=10, color=GREEN).next_to(punto, direction, buff=0.1)
            esquinas.add(VGroup(punto, label))
        
        # Bordes
        bordes = VGroup()
        for direction, name in [(UP, "UP"), (DOWN, "DOWN"), (LEFT, "LEFT"), (RIGHT, "RIGHT")]:
            punto = Dot(cuadrado.get_edge_center(direction), color=YELLOW)
            label = Text(name, font_size=10, color=YELLOW).next_to(punto, direction, buff=0.1)
            bordes.add(VGroup(punto, label))
        
        self.play(Create(centro), Write(label_centro))
        self.play(*[Create(e) for e in esquinas])
        self.play(*[Create(b) for b in bordes])
        
        self.wait()
````

## 7. ValueTracker - Seguimiento de Valores

`ValueTracker` es una herramienta fundamental para crear animaciones basadas en valores numéricos que cambian con el tiempo.

### Uso básico:
```python
tracker = ValueTracker(valor_inicial)
tracker.get_value()  # Obtener valor actual
tracker.set_value(nuevo_valor)  # Establecer valor

# Animar el cambio de valor
self.play(tracker.animate.set_value(valor_final))
```

### Uso con updaters:
Se combina con `always_redraw()` para crear objetos que se actualizan automáticamente.

````python
%%manim -qm -v WARNING ValueTrackerDemo

class ValueTrackerDemo(Scene):
    def construct(self):
        titulo = Text("ValueTracker - Seguimiento de Valores", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ValueTracker
        numero = ValueTracker(0)
        
        # Display que se actualiza con el tracker
        display = always_redraw(
            lambda: Text(f"Valor: {numero.get_value():.2f}", 
                        font_size=48, color=BLUE).shift(UP)
        )
        
        # Barra de progreso que se actualiza
        barra_fondo = Rectangle(width=8, height=0.5, color=GREY, fill_opacity=0.3)
        barra_fondo.shift(DOWN)
        
        barra_progreso = always_redraw(
            lambda: Rectangle(
                width=8 * (numero.get_value() / 100),
                height=0.5,
                color=GREEN,
                fill_opacity=0.8
            ).move_to(barra_fondo.get_left(), aligned_edge=LEFT)
        )
        
        self.play(Create(display), Create(barra_fondo))
        self.add(barra_progreso)
        
        # Animar el tracker
        info = Text("Animando ValueTracker de 0 a 100", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.play(numero.animate.set_value(100), run_time=3)
        
        self.play(FadeOut(info))
        
        # Volver a 0
        info2 = Text("De vuelta a 0", font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(numero.animate.set_value(0), run_time=2)
        
        # Ir a 50 con rate_func
        self.play(FadeOut(info2))
        info3 = Text("A 50 con rate_func=there_and_back", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3))
        self.play(
            numero.animate.set_value(100), 
            rate_func=there_and_back, 
            run_time=3
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING ValueTrackerGraph

class ValueTrackerGraph(Scene):
    def construct(self):
        titulo = Text("ValueTracker con Gráficas", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        ejes = Axes(
            x_range=[0, 2*PI, PI/2],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=4,
            axis_config={"include_tip": True}
        )
        
        labels = ejes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(ejes), Write(labels))
        
        # ValueTracker para el parámetro
        t = ValueTracker(0)
        
        # Curva que se dibuja según el valor de t
        curva = always_redraw(
            lambda: ejes.plot(
                lambda x: np.sin(x),
                x_range=[0, t.get_value()],
                color=BLUE
            )
        )
        
        # Punto que sigue la curva
        punto = always_redraw(
            lambda: Dot(
                ejes.c2p(t.get_value(), np.sin(t.get_value())),
                color=RED
            )
        )
        
        # Etiqueta del punto
        label_punto = always_redraw(
            lambda: MathTex(
                f"({t.get_value():.2f}, {np.sin(t.get_value()):.2f})",
                font_size=24
            ).next_to(punto, UR, buff=0.1)
        )
        
        self.add(curva, punto, label_punto)
        
        info = Text("Dibujando sin(x) con ValueTracker", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        # Animar el dibujo
        self.play(t.animate.set_value(2*PI), run_time=4)
        
        # Volver
        self.play(t.animate.set_value(0), run_time=2)
        
        # Ir más rápido
        self.play(t.animate.set_value(2*PI), run_time=2, rate_func=rush_into)
        
        self.wait()
````

## 8. Funciones Matemáticas Útiles

Manim incluye varias funciones matemáticas útiles además de las de NumPy.

### Funciones de interpolación:
- `interpolate(a, b, alpha)` - Interpola entre dos valores
- `inverse_interpolate(a, b, x)` - Obtiene alpha dado x
- `match_interpolate(a, b, c, d, x)` - Mapea x de [a,b] a [c,d]

### Funciones de suavizado:
- `smooth(t)` - Función suave S-curve
- `rush_into(t)` - Acelera al final
- `rush_from(t)` - Desacelera desde inicio

### Funciones geométricas:
- `midpoint(p1, p2)` - Punto medio entre dos puntos
- `line_intersection(line1, line2)` - Intersección de líneas

````python
%%manim -qm -v WARNING MathFunctionsDemo

class MathFunctionsDemo(Scene):
    def construct(self):
        titulo = Text("Funciones Matemáticas Útiles", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        # Demostrar interpolate
        subtitulo1 = Text("interpolate(a, b, alpha)", font_size=22, color=BLUE).shift(UP * 2)
        self.play(Write(subtitulo1))
        
        # Crear dos puntos
        punto_a = Dot(LEFT * 4, color=RED)
        punto_b = Dot(RIGHT * 4, color=GREEN)
        label_a = Text("a", font_size=16).next_to(punto_a, DOWN)
        label_b = Text("b", font_size=16).next_to(punto_b, DOWN)
        
        linea = Line(LEFT * 4, RIGHT * 4, color=GREY)
        
        self.play(Create(linea), Create(punto_a), Create(punto_b), 
                 Write(label_a), Write(label_b))
        
        # Mostrar puntos interpolados
        alphas = [0.25, 0.5, 0.75]
        puntos_interp = VGroup()
        
        for alpha in alphas:
            pos = interpolate(LEFT * 4, RIGHT * 4, alpha)
            punto = Dot(pos, color=YELLOW)
            label = Text(f"α={alpha}", font_size=12).next_to(punto, UP, buff=0.1)
            puntos_interp.add(VGroup(punto, label))
        
        self.play(*[Create(p) for p in puntos_interp])
        
        info1 = Text("interpolate mezcla dos valores según alpha ∈ [0,1]", 
                    font_size=14, color=WHITE).to_edge(DOWN)
        self.play(Write(info1))
        self.wait()
        
        self.play(*[FadeOut(m) for m in [subtitulo1, linea, punto_a, punto_b, 
                                         label_a, label_b, puntos_interp, info1]])
        
        # Demostrar midpoint
        subtitulo2 = Text("midpoint(p1, p2)", font_size=22, color=BLUE).shift(UP * 2)
        self.play(Write(subtitulo2))
        
        p1 = np.array([-3, -1, 0])
        p2 = np.array([3, 1, 0])
        
        dot1 = Dot(p1, color=RED)
        dot2 = Dot(p2, color=GREEN)
        linea2 = Line(p1, p2, color=GREY)
        
        # Calcular punto medio
        pm = midpoint(p1, p2)
        dot_mid = Dot(pm, color=YELLOW)
        label_mid = Text("midpoint", font_size=14, color=YELLOW).next_to(dot_mid, UP)
        
        self.play(Create(linea2), Create(dot1), Create(dot2))
        self.play(Create(dot_mid), Write(label_mid))
        
        info2 = Text(f"midpoint = ({pm[0]:.1f}, {pm[1]:.1f})", 
                    font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info2))
        
        self.wait()
````

## 9. Ejercicios Prácticos

Practica lo aprendido con estos ejercicios.

````python
%%manim -qm -v WARNING Ejercicio1_RateFuncPersonalizada

# Ejercicio 1: Crear una animación con rate function personalizada
# que simule el movimiento de un péndulo

class Ejercicio1_RateFuncPersonalizada(Scene):
    def construct(self):
        titulo = Text("Ejercicio 1: Péndulo con Rate Function", font_size=26).to_edge(UP)
        self.play(Write(titulo))
        
        # Rate function tipo péndulo (oscilación amortiguada)
        def pendulum(t):
            return np.sin(t * 4 * PI) * (1 - t) * 0.5 + 0.5
        
        # Crear el péndulo
        pivote = Dot(UP * 2.5, color=WHITE)
        cuerda = always_redraw(
            lambda: Line(pivote.get_center(), bola.get_center(), color=GREY)
        )
        bola = Circle(radius=0.3, color=YELLOW, fill_opacity=1)
        bola.move_to(DOWN * 0.5 + LEFT * 2)
        
        self.play(Create(pivote))
        self.add(cuerda)
        self.play(Create(bola))
        
        info = Text("Rate function: péndulo amortiguado", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        # Animar el péndulo
        self.play(
            bola.animate.move_to(DOWN * 0.5 + RIGHT * 2),
            rate_func=pendulum,
            run_time=4
        )
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio2_GradienteAnimado

# Ejercicio 2: Crear un gradiente de colores animado que rota

class Ejercicio2_GradienteAnimado(Scene):
    def construct(self):
        titulo = Text("Ejercicio 2: Rueda de Colores", font_size=26).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear rueda de colores usando sectores
        n_sectores = 12
        colores = [RED, ORANGE, YELLOW, GREEN, TEAL, BLUE, 
                  PURPLE, PINK, RED, ORANGE, YELLOW, GREEN]
        
        sectores = VGroup()
        for i in range(n_sectores):
            angulo_inicio = i * TAU / n_sectores
            angulo_fin = (i + 1) * TAU / n_sectores
            
            sector = AnnularSector(
                inner_radius=0.5,
                outer_radius=2,
                angle=TAU / n_sectores,
                start_angle=angulo_inicio,
                color=colores[i % len(colores)],
                fill_opacity=0.8
            )
            sectores.add(sector)
        
        self.play(LaggedStart(*[GrowFromCenter(s) for s in sectores], lag_ratio=0.1))
        
        info = Text("Rotación continua de la rueda", 
                   font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info))
        
        # Rotar la rueda
        self.play(Rotate(sectores, TAU, run_time=4, rate_func=linear))
        
        # Cambio de colores
        self.play(FadeOut(info))
        info2 = Text("Interpolación de colores", 
                    font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info2))
        
        # Animar cambio de colores a monocromático
        animaciones = [
            sector.animate.set_fill(interpolate_color(sector.get_fill_color(), WHITE, 0.5))
            for sector in sectores
        ]
        self.play(*animaciones)
        
        # Volver a colores
        animaciones2 = [
            sectores[i].animate.set_fill(colores[i % len(colores)])
            for i in range(n_sectores)
        ]
        self.play(*animaciones2)
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio3_ContadorAnimado

# Ejercicio 3: Contador animado usando ValueTracker

class Ejercicio3_ContadorAnimado(Scene):
    def construct(self):
        titulo = Text("Ejercicio 3: Contador Animado", font_size=26).to_edge(UP)
        self.play(Write(titulo))
        
        # ValueTracker para el contador
        contador = ValueTracker(0)
        
        # Display del contador
        display = always_redraw(
            lambda: Text(
                f"{int(contador.get_value()):03d}",
                font_size=120,
                color=BLUE
            )
        )
        
        # Círculo indicador
        circulo_fondo = Circle(radius=2.5, color=GREY, stroke_width=10)
        
        arco_progreso = always_redraw(
            lambda: Arc(
                radius=2.5,
                start_angle=PI/2,
                angle=-TAU * (contador.get_value() / 100),
                color=GREEN,
                stroke_width=10
            )
        )
        
        self.play(Create(circulo_fondo))
        self.add(arco_progreso)
        self.play(Write(display))
        
        info = Text("Contando de 0 a 100", font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info))
        
        # Animar el contador
        self.play(contador.animate.set_value(100), run_time=5, rate_func=linear)
        
        # Efecto de celebración
        self.play(FadeOut(info))
        
        estrellas = VGroup(*[
            Star(5, color=YELLOW, fill_opacity=0.8).scale(0.3).move_to(
                np.array([np.random.uniform(-5, 5), np.random.uniform(-2, 2), 0])
            )
            for _ in range(10)
        ])
        
        self.play(
            LaggedStart(*[GrowFromCenter(s) for s in estrellas], lag_ratio=0.1),
            Flash(display, color=YELLOW, line_length=0.5)
        )
        
        self.wait()
````
