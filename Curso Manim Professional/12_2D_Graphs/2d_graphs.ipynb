## 1. Axes - Sistema de Ejes Coordenados

`Axes` es la clase principal para crear sistemas de coordenadas en 2D.

### Parámetros principales:
- **`x_range`**: [min, max, step] - Rango del eje X
- **`y_range`**: [min, max, step] - Rango del eje Y
- **`x_length`**: Longitud visual del eje X
- **`y_length`**: Longitud visual del eje Y
- **`axis_config`**: Configuración de los ejes
- **`x_axis_config`** / **`y_axis_config`**: Configuración individual

### Métodos importantes:
- `axes.plot(función)` - Graficar una función
- `axes.coords_to_point(x, y)` - Convertir coordenadas a posición
- `axes.point_to_coords(punto)` - Convertir posición a coordenadas
- `axes.get_axis_labels()` - Obtener etiquetas de ejes

````python
%%manim -qm -v WARNING AxesBasico

class AxesBasico(Scene):
    def construct(self):
        titulo = Text("Axes - Sistema de Coordenadas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes básicos
        axes = Axes(
            x_range=[-5, 5, 1],  # [min, max, step]
            y_range=[-3, 3, 1],
            x_length=10,
            y_length=6,
            axis_config={
                "color": BLUE,
                "include_tip": True,
                "include_numbers": True
            }
        )
        
        # Etiquetas de ejes
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(axes), run_time=2)
        self.play(Write(labels))
        
        # Mostrar información
        info = Text("Axes con x_range=[-5,5], y_range=[-3,3]", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

````python
%%manim -qm -v WARNING AxesPersonalizado

class AxesPersonalizado(Scene):
    def construct(self):
        titulo = Text("Axes Personalizado", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Ejes con estilo personalizado
        axes = Axes(
            x_range=[-4, 4, 0.5],
            y_range=[-2, 2, 0.5],
            x_length=8,
            y_length=4,
            x_axis_config={
                "color": RED,
                "include_tip": True,
                "numbers_to_include": [-4, -2, 0, 2, 4],
                "font_size": 20
            },
            y_axis_config={
                "color": GREEN,
                "include_tip": True,
                "numbers_to_include": [-2, -1, 0, 1, 2],
                "font_size": 20
            },
            tips=True
        )
        
        # Etiquetas personalizadas
        x_label = MathTex("x").next_to(axes.x_axis, RIGHT)
        y_label = MathTex("f(x)").next_to(axes.y_axis, UP)
        
        self.play(Create(axes), run_time=2)
        self.play(Write(x_label), Write(y_label))
        
        # Añadir grid
        grid = axes.get_graph(lambda x: 0, color=GREY, stroke_opacity=0)
        
        info = Text("Ejes con colores y números personalizados", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

## 2. NumberPlane - Plano Numérico con Grid

`NumberPlane` extiende `Axes` añadiendo una cuadrícula de fondo, ideal para visualizar coordenadas.

### Parámetros adicionales:
- **`background_line_style`**: Estilo de las líneas del grid
- **`faded_line_style`**: Estilo de líneas secundarias
- **`faded_line_ratio`**: Proporción de líneas atenuadas

````python
%%manim -qm -v WARNING NumberPlaneDemo

class NumberPlaneDemo(Scene):
    def construct(self):
        titulo = Text("NumberPlane - Plano con Grid", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # NumberPlane básico
        plano = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-4, 4, 1],
            background_line_style={
                "stroke_color": BLUE,
                "stroke_width": 1,
                "stroke_opacity": 0.6
            }
        )
        
        self.play(Create(plano), run_time=2)
        
        # Mostrar algunos puntos
        puntos_coords = [(2, 2), (-3, 1), (4, -2), (-2, -3)]
        puntos = VGroup()
        
        for x, y in puntos_coords:
            punto = Dot(plano.coords_to_point(x, y), color=RED)
            label = Text(f"({x},{y})", font_size=12, color=YELLOW)
            label.next_to(punto, UR, buff=0.1)
            puntos.add(VGroup(punto, label))
        
        info = Text("coords_to_point(x, y) convierte coordenadas a posición", 
                   font_size=14, color=WHITE).to_edge(DOWN)
        self.play(Write(info))
        
        for punto in puntos:
            self.play(Create(punto), run_time=0.5)
        
        self.wait()
````

## 3. plot() - Graficar Funciones

El método `plot()` de Axes permite graficar funciones matemáticas.

### Sintaxis:
```python
grafica = axes.plot(lambda x: función, color=COLOR, x_range=[min, max])
```

### Parámetros:
- **`function`**: Función lambda o definida
- **`color`**: Color de la gráfica
- **`x_range`**: Rango de x para la gráfica (opcional)
- **`discontinuities`**: Lista de puntos de discontinuidad
- **`dt`**: Paso de discretización

````python
%%manim -qm -v WARNING PlotBasico

class PlotBasico(Scene):
    def construct(self):
        titulo = Text("plot() - Graficar Funciones", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-4, 4, 1],
            y_range=[-2, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 20}
        )
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(axes), Write(labels))
        
        # Graficar f(x) = x²
        grafica = axes.plot(lambda x: x**2, color=BLUE, x_range=[-2, 2])
        
        # Etiqueta de la función
        func_label = MathTex("f(x) = x^2", color=BLUE).to_corner(UR)
        
        info = Text("axes.plot(lambda x: x**2)", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.play(Create(grafica), Write(func_label), run_time=2)
        
        self.wait()
````

````python
%%manim -qm -v WARNING PlotVariasFunciones

class PlotVariasFunciones(Scene):
    def construct(self):
        titulo = Text("Varias Funciones en un Gráfico", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-PI, PI, PI/2],
            y_range=[-1.5, 1.5, 0.5],
            x_length=10,
            y_length=5,
            axis_config={"include_tip": True}
        )
        
        # Etiquetas personalizadas para PI
        x_labels = axes.get_x_axis().add_labels({
            -PI: MathTex("-\\pi"),
            -PI/2: MathTex("-\\frac{\\pi}{2}"),
            PI/2: MathTex("\\frac{\\pi}{2}"),
            PI: MathTex("\\pi")
        })
        
        self.play(Create(axes))
        
        # Graficar sen(x)
        sin_graph = axes.plot(lambda x: np.sin(x), color=BLUE)
        sin_label = MathTex("\\sin(x)", color=BLUE, font_size=28).shift(UP * 2.5 + LEFT * 4)
        
        # Graficar cos(x)
        cos_graph = axes.plot(lambda x: np.cos(x), color=RED)
        cos_label = MathTex("\\cos(x)", color=RED, font_size=28).shift(UP * 2.5 + RIGHT * 4)
        
        # Animar las gráficas
        self.play(Create(sin_graph), Write(sin_label), run_time=2)
        self.play(Create(cos_graph), Write(cos_label), run_time=2)
        
        info = Text("Funciones trigonométricas", 
                   font_size=16, color=WHITE).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

## 4. ParametricFunction - Funciones Paramétricas

Las funciones paramétricas permiten graficar curvas donde x e y dependen de un parámetro t.

### Sintaxis:
```python
curva = ParametricFunction(
    lambda t: np.array([x(t), y(t), 0]),
    t_range=[t_min, t_max],
    color=COLOR
)
```

### Ejemplos clásicos:
- **Círculo**: x = cos(t), y = sin(t)
- **Espiral**: x = t·cos(t), y = t·sin(t)
- **Lissajous**: x = sin(at), y = sin(bt)

````python
%%manim -qm -v WARNING ParametricFunctionDemo

class ParametricFunctionDemo(Scene):
    def construct(self):
        titulo = Text("Funciones Paramétricas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Espiral
        espiral = ParametricFunction(
            lambda t: np.array([
                t * np.cos(4 * t) / 3,
                t * np.sin(4 * t) / 3,
                0
            ]),
            t_range=[0, 3 * PI],
            color=BLUE
        )
        
        label_espiral = MathTex(
            "x = t\\cos(4t), \\quad y = t\\sin(4t)",
            font_size=24
        ).to_edge(DOWN)
        
        self.play(Create(espiral), Write(label_espiral), run_time=3)
        self.wait()
        
        self.play(FadeOut(espiral), FadeOut(label_espiral))
        
        # Curva de Lissajous
        lissajous = ParametricFunction(
            lambda t: np.array([
                2 * np.sin(3 * t),
                2 * np.sin(4 * t),
                0
            ]),
            t_range=[0, TAU],
            color=GREEN
        )
        
        label_lissajous = MathTex(
            "x = 2\\sin(3t), \\quad y = 2\\sin(4t)",
            font_size=24
        ).to_edge(DOWN)
        
        self.play(Create(lissajous), Write(label_lissajous), run_time=3)
        self.wait()
        
        self.play(FadeOut(lissajous), FadeOut(label_lissajous))
        
        # Cardioide
        cardioide = ParametricFunction(
            lambda t: np.array([
                1.5 * (1 - np.cos(t)) * np.cos(t),
                1.5 * (1 - np.cos(t)) * np.sin(t),
                0
            ]),
            t_range=[0, TAU],
            color=RED
        )
        
        label_cardioide = Text("Cardioide", font_size=20, color=RED).to_edge(DOWN)
        
        self.play(Create(cardioide), Write(label_cardioide), run_time=3)
        
        self.wait()
````

## 5. get_area() - Áreas Bajo Curvas

El método `get_area()` permite visualizar el área bajo una curva.

### Sintaxis:
```python
area = axes.get_area(
    graph,
    x_range=[a, b],
    color=COLOR,
    opacity=0.5
)
```

### Parámetros:
- **`graph`**: La gráfica sobre la cual calcular el área
- **`x_range`**: Rango [a, b] del área
- **`color`**: Color del área
- **`opacity`**: Transparencia del relleno
- **`bounded_graph`**: Gráfica inferior (para área entre curvas)

````python
%%manim -qm -v WARNING AreaBajoCurva

class AreaBajoCurva(Scene):
    def construct(self):
        titulo = Text("Área Bajo la Curva", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-1, 5, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 20}
        )
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(axes), Write(labels))
        
        # Graficar f(x) = x²/4
        func = lambda x: x**2 / 4
        grafica = axes.plot(func, color=BLUE, x_range=[0, 4])
        func_label = MathTex("f(x) = \\frac{x^2}{4}", color=BLUE).to_corner(UR)
        
        self.play(Create(grafica), Write(func_label))
        
        # Área bajo la curva de 1 a 3
        area = axes.get_area(grafica, x_range=[1, 3], color=GREEN, opacity=0.5)
        
        # Etiquetas de los límites
        a_label = MathTex("a=1", font_size=24).next_to(axes.c2p(1, 0), DOWN)
        b_label = MathTex("b=3", font_size=24).next_to(axes.c2p(3, 0), DOWN)
        
        info = MathTex(
            "\\int_1^3 \\frac{x^2}{4} dx",
            font_size=28, color=GREEN
        ).to_edge(DOWN)
        
        self.play(Write(a_label), Write(b_label))
        self.play(FadeIn(area), Write(info), run_time=2)
        
        self.wait()
````

````python
%%manim -qm -v WARNING AreaEntreCurvas

class AreaEntreCurvas(Scene):
    def construct(self):
        titulo = Text("Área Entre Curvas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-1, 4, 1],
            y_range=[-1, 5, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 20}
        )
        
        self.play(Create(axes))
        
        # Función superior f(x) = x
        func_superior = lambda x: x
        grafica_sup = axes.plot(func_superior, color=BLUE, x_range=[0, 3.5])
        label_sup = MathTex("f(x) = x", color=BLUE, font_size=24).shift(UP * 2 + RIGHT * 3)
        
        # Función inferior g(x) = x²/4
        func_inferior = lambda x: x**2 / 4
        grafica_inf = axes.plot(func_inferior, color=RED, x_range=[0, 3.5])
        label_inf = MathTex("g(x) = \\frac{x^2}{4}", color=RED, font_size=24).shift(DOWN + RIGHT * 3)
        
        self.play(Create(grafica_sup), Write(label_sup))
        self.play(Create(grafica_inf), Write(label_inf))
        
        # Área entre las curvas de 0 a 3
        area = axes.get_area(
            grafica_sup,
            x_range=[0, 3],
            bounded_graph=grafica_inf,
            color=YELLOW,
            opacity=0.5
        )
        
        info = MathTex(
            "\\int_0^3 \\left( x - \\frac{x^2}{4} \\right) dx",
            font_size=24, color=YELLOW
        ).to_edge(DOWN)
        
        self.play(FadeIn(area), Write(info), run_time=2)
        
        self.wait()
````

## 6. get_riemann_rectangles() - Rectángulos de Riemann

Visualiza la aproximación del área mediante rectángulos de Riemann.

### Sintaxis:
```python
rectangles = axes.get_riemann_rectangles(
    graph,
    x_range=[a, b],
    dx=ancho,
    input_sample_type="left"  # "left", "right", "center"
)
```

### Tipos de muestreo:
- **"left"**: Altura en el borde izquierdo
- **"right"**: Altura en el borde derecho
- **"center"**: Altura en el centro

````python
%%manim -qm -v WARNING RiemannRectangles

class RiemannRectangles(Scene):
    def construct(self):
        titulo = Text("Rectángulos de Riemann", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-1, 5, 1],
            x_length=7,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        
        self.play(Create(axes))
        
        # Graficar f(x) = x²/4
        func = lambda x: x**2 / 4
        grafica = axes.plot(func, color=BLUE, x_range=[0, 4])
        func_label = MathTex("f(x) = \\frac{x^2}{4}", color=BLUE, font_size=24).to_corner(UR)
        
        self.play(Create(grafica), Write(func_label))
        
        # Rectángulos con diferentes dx
        dx_valores = [1, 0.5, 0.25, 0.1]
        
        rectangulos_anteriores = None
        
        for dx in dx_valores:
            rectangulos = axes.get_riemann_rectangles(
                grafica,
                x_range=[0, 4],
                dx=dx,
                input_sample_type="left",
                color=[BLUE, GREEN],
                fill_opacity=0.5,
                stroke_width=1
            )
            
            info = Text(f"dx = {dx}, n = {int(4/dx)} rectángulos", 
                       font_size=18, color=YELLOW).to_edge(DOWN)
            
            if rectangulos_anteriores is None:
                self.play(Create(rectangulos), Write(info))
            else:
                self.play(
                    Transform(rectangulos_anteriores, rectangulos),
                    Transform(info_anterior, info)
                )
                rectangulos = rectangulos_anteriores
            
            rectangulos_anteriores = rectangulos
            info_anterior = info
            self.wait(0.5)
        
        # Mostrar el área real
        area = axes.get_area(grafica, x_range=[0, 4], color=GREEN, opacity=0.3)
        self.play(FadeIn(area))
        
        conclusion = Text("Más rectángulos → Mejor aproximación", 
                         font_size=16, color=WHITE).shift(DOWN * 2.5)
        self.play(Write(conclusion))
        
        self.wait()
````

## 7. Líneas Tangentes y Secantes

Manim permite crear líneas tangentes a curvas en puntos específicos.

### Método:
```python
tangente = axes.get_secant_slope_group(
    x=punto_x,
    graph=grafica,
    dx=0.01,  # Pequeño para tangente, grande para secante
    secant_line_color=COLOR
)
```

### Alternativa manual:
Calcular la pendiente usando la derivada y crear una línea.

````python
%%manim -qm -v WARNING TangentLineDemo

class TangentLineDemo(Scene):
    def construct(self):
        titulo = Text("Línea Tangente", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-1, 5, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        
        self.play(Create(axes))
        
        # Función f(x) = x²/4
        func = lambda x: x**2 / 4
        func_derivada = lambda x: x / 2  # f'(x) = x/2
        
        grafica = axes.plot(func, color=BLUE, x_range=[0, 4])
        func_label = MathTex("f(x) = \\frac{x^2}{4}", color=BLUE, font_size=24).to_corner(UR)
        
        self.play(Create(grafica), Write(func_label))
        
        # ValueTracker para el punto x
        x_tracker = ValueTracker(1)
        
        # Punto en la curva
        punto = always_redraw(
            lambda: Dot(
                axes.c2p(x_tracker.get_value(), func(x_tracker.get_value())),
                color=RED
            )
        )
        
        # Línea tangente
        def get_tangent_line():
            x = x_tracker.get_value()
            y = func(x)
            m = func_derivada(x)
            
            # Línea: y - y0 = m(x - x0)
            x1, x2 = x - 1.5, x + 1.5
            y1 = y + m * (x1 - x)
            y2 = y + m * (x2 - x)
            
            return Line(
                axes.c2p(x1, y1),
                axes.c2p(x2, y2),
                color=YELLOW
            )
        
        tangente = always_redraw(get_tangent_line)
        
        # Label de pendiente
        pendiente_label = always_redraw(
            lambda: MathTex(
                f"m = f'({x_tracker.get_value():.1f}) = {func_derivada(x_tracker.get_value()):.2f}",
                font_size=24, color=YELLOW
            ).to_edge(DOWN)
        )
        
        self.play(Create(punto), Create(tangente), Write(pendiente_label))
        
        # Mover el punto a lo largo de la curva
        self.play(x_tracker.animate.set_value(3), run_time=4)
        self.play(x_tracker.animate.set_value(0.5), run_time=3)
        self.play(x_tracker.animate.set_value(2), run_time=2)
        
        self.wait()
````

## 8. Coordenadas y Puntos en Gráficas

Métodos útiles para trabajar con coordenadas en gráficas.

### Conversión de coordenadas:
- `axes.c2p(x, y)` - Coordenadas a punto (coords_to_point)
- `axes.p2c(punto)` - Punto a coordenadas (point_to_coords)
- `axes.i2gp(x, graph)` - Input to graph point (punto en la gráfica)

### Crear puntos en gráficas:
```python
punto = Dot(axes.i2gp(x, grafica), color=RED)
```

````python
%%manim -qm -v WARNING CoordinatesDemo

class CoordinatesDemo(Scene):
    def construct(self):
        titulo = Text("Coordenadas y Puntos", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-1, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        
        self.play(Create(axes))
        
        # Graficar función
        func = lambda x: np.sqrt(x)
        grafica = axes.plot(func, color=BLUE, x_range=[0, 4.5])
        func_label = MathTex("f(x) = \\sqrt{x}", color=BLUE, font_size=24).to_corner(UR)
        
        self.play(Create(grafica), Write(func_label))
        
        # Marcar puntos específicos usando i2gp
        puntos_x = [1, 2, 3, 4]
        
        for x in puntos_x:
            # Punto en la gráfica
            punto = Dot(axes.i2gp(x, grafica), color=RED)
            
            # Líneas punteadas a los ejes
            linea_v = DashedLine(
                axes.c2p(x, 0),
                axes.i2gp(x, grafica),
                color=GREY
            )
            linea_h = DashedLine(
                axes.c2p(0, func(x)),
                axes.i2gp(x, grafica),
                color=GREY
            )
            
            # Etiqueta del punto
            y_val = func(x)
            label = MathTex(f"({x}, {y_val:.2f})", font_size=16).next_to(punto, UR, buff=0.1)
            
            self.play(
                Create(linea_v), Create(linea_h),
                Create(punto), Write(label),
                run_time=0.8
            )
        
        info = Text("i2gp(x, graph) obtiene el punto (x, f(x)) en la gráfica", 
                   font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

## 9. Animaciones de Gráficas

Las gráficas se pueden animar de varias formas interesantes.

### Animaciones disponibles:
- **`Create(graph)`**: Dibuja la gráfica progresivamente
- **`Transform(graph1, graph2)`**: Transforma una gráfica en otra
- **`MoveAlongPath(dot, graph)`**: Mueve un punto a lo largo de la curva

### Técnica avanzada:
Usar `ValueTracker` para animar parámetros de funciones.

````python
%%manim -qm -v WARNING GraphAnimations

class GraphAnimations(Scene):
    def construct(self):
        titulo = Text("Animaciones de Gráficas", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-4, 4, 1],
            y_range=[-2, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        
        self.play(Create(axes))
        
        # Gráfica 1: x²
        graph1 = axes.plot(lambda x: x**2, color=BLUE, x_range=[-2, 2])
        label1 = MathTex("f(x) = x^2", color=BLUE, font_size=24).to_corner(UR)
        
        info1 = Text("Create(graph) - Dibujo progresivo", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info1))
        self.play(Create(graph1), Write(label1), run_time=2)
        
        self.play(FadeOut(info1))
        
        # Transform a otra función
        graph2 = axes.plot(lambda x: -x**2 + 3, color=RED, x_range=[-2, 2])
        label2 = MathTex("g(x) = -x^2 + 3", color=RED, font_size=24).to_corner(UR)
        
        info2 = Text("Transform(graph1, graph2) - Transformación", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info2))
        self.play(Transform(graph1, graph2), Transform(label1, label2), run_time=2)
        
        self.play(FadeOut(info2))
        
        # MoveAlongPath
        punto = Dot(color=GREEN).move_to(axes.c2p(-2, -1))
        
        info3 = Text("MoveAlongPath - Punto siguiendo la curva", 
                    font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info3), Create(punto))
        self.play(MoveAlongPath(punto, graph1), run_time=3)
        
        self.wait()
````

````python
%%manim -qm -v WARNING AnimatedParameter

class AnimatedParameter(Scene):
    def construct(self):
        titulo = Text("Animación de Parámetros", font_size=30).to_edge(UP)
        self.play(Write(titulo))
        
        # Crear ejes
        axes = Axes(
            x_range=[-4, 4, 1],
            y_range=[-3, 3, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        
        self.play(Create(axes))
        
        # ValueTracker para el parámetro 'a' en f(x) = a * sin(x)
        a = ValueTracker(1)
        
        # Gráfica que se actualiza con el parámetro
        grafica = always_redraw(
            lambda: axes.plot(
                lambda x: a.get_value() * np.sin(x),
                color=BLUE,
                x_range=[-4, 4]
            )
        )
        
        # Label que muestra el valor actual de 'a'
        label = always_redraw(
            lambda: MathTex(
                f"f(x) = {a.get_value():.1f} \\sin(x)",
                color=BLUE, font_size=28
            ).to_corner(UR)
        )
        
        self.play(Create(grafica), Write(label))
        
        info = Text("Animando el parámetro 'a'", 
                   font_size=16, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        # Animar el parámetro
        self.play(a.animate.set_value(2), run_time=2)
        self.play(a.animate.set_value(0.5), run_time=2)
        self.play(a.animate.set_value(-1.5), run_time=2)
        self.play(a.animate.set_value(1), run_time=2)
        
        self.wait()
````

## 10. Ejercicios Prácticos

````python
%%manim -qm -v WARNING Ejercicio1_Derivada

# Ejercicio 1: Visualización de Derivada
# Mostrar una función y su derivada superpuestas

class Ejercicio1_Derivada(Scene):
    def construct(self):
        titulo = Text("Ejercicio 1: f(x) y f'(x)", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        axes = Axes(
            x_range=[-3, 3, 1],
            y_range=[-2, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 16}
        )
        labels = axes.get_axis_labels(x_label="x", y_label="y")
        
        self.play(Create(axes), Write(labels))
        
        # f(x) = x³/3 - x
        func = axes.plot(lambda x: x**3/3 - x, color=BLUE, x_range=[-2.5, 2.5])
        func_label = MathTex("f(x) = \\frac{x^3}{3} - x", color=BLUE, font_size=22)
        func_label.to_corner(UL).shift(DOWN * 0.5)
        
        # f'(x) = x² - 1
        derivada = axes.plot(lambda x: x**2 - 1, color=RED, x_range=[-2.5, 2.5])
        der_label = MathTex("f'(x) = x^2 - 1", color=RED, font_size=22)
        der_label.next_to(func_label, DOWN, aligned_edge=LEFT)
        
        self.play(Create(func), Write(func_label), run_time=2)
        self.play(Create(derivada), Write(der_label), run_time=2)
        
        # Marcar puntos críticos (donde f'(x) = 0)
        puntos_criticos = [-1, 1]
        for x in puntos_criticos:
            punto = Dot(axes.i2gp(x, func), color=YELLOW)
            linea = DashedLine(
                axes.c2p(x, -2),
                axes.c2p(x, 4),
                color=YELLOW, stroke_opacity=0.5
            )
            self.play(Create(linea), Create(punto))
        
        info = Text("Puntos donde f'(x) = 0 son máximos/mínimos de f(x)", 
                   font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info))
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio2_Integral

# Ejercicio 2: Animación de Integral como Área Acumulada

class Ejercicio2_Integral(Scene):
    def construct(self):
        titulo = Text("Ejercicio 2: Integral como Área", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        axes = Axes(
            x_range=[-1, 5, 1],
            y_range=[-1, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 16}
        )
        
        self.play(Create(axes))
        
        # f(x) = x
        func = lambda x: x
        grafica = axes.plot(func, color=BLUE, x_range=[0, 4])
        func_label = MathTex("f(x) = x", color=BLUE, font_size=24).to_corner(UR)
        
        self.play(Create(grafica), Write(func_label))
        
        # Tracker para el límite superior
        b = ValueTracker(0.1)
        
        # Área que crece
        area = always_redraw(
            lambda: axes.get_area(
                grafica,
                x_range=[0, b.get_value()],
                color=GREEN,
                opacity=0.5
            )
        )
        
        # Label del área
        area_label = always_redraw(
            lambda: MathTex(
                f"\\int_0^{{{b.get_value():.1f}}} x \\, dx = {b.get_value()**2/2:.2f}",
                font_size=24, color=GREEN
            ).to_edge(DOWN)
        )
        
        self.add(area)
        self.play(Write(area_label))
        
        # Animar el crecimiento del área
        self.play(b.animate.set_value(4), run_time=5, rate_func=linear)
        
        # Resultado final
        resultado = MathTex(
            "\\int_0^4 x \\, dx = \\frac{x^2}{2}\\Big|_0^4 = 8",
            font_size=28, color=YELLOW
        ).shift(DOWN * 2)
        
        self.play(Write(resultado))
        
        self.wait()
````

````python
%%manim -qm -v WARNING Ejercicio3_Transformaciones

# Ejercicio 3: Transformaciones de funciones
# Mostrar cómo f(x), f(x+a), f(x)+b, af(x) transforman la gráfica

class Ejercicio3_Transformaciones(Scene):
    def construct(self):
        titulo = Text("Ejercicio 3: Transformaciones", font_size=28).to_edge(UP)
        self.play(Write(titulo))
        
        axes = Axes(
            x_range=[-4, 4, 1],
            y_range=[-2, 4, 1],
            x_length=8,
            y_length=5,
            axis_config={"include_numbers": True, "font_size": 16}
        )
        
        self.play(Create(axes))
        
        # f(x) = x² original
        f_original = axes.plot(lambda x: x**2, color=WHITE, x_range=[-2, 2])
        label_original = MathTex("f(x) = x^2", color=WHITE, font_size=20)
        label_original.to_corner(UL).shift(DOWN * 0.5)
        
        self.play(Create(f_original), Write(label_original))
        
        # f(x) + 1 (traslación vertical)
        f_vertical = axes.plot(lambda x: x**2 + 1, color=RED, x_range=[-2, 2])
        label_v = MathTex("f(x) + 1", color=RED, font_size=18).next_to(label_original, DOWN, aligned_edge=LEFT)
        
        info1 = Text("f(x) + 1: Traslación vertical", font_size=14, color=RED).to_edge(DOWN)
        self.play(Write(info1))
        self.play(Create(f_vertical), Write(label_v))
        self.wait(0.5)
        
        self.play(FadeOut(f_vertical), FadeOut(label_v), FadeOut(info1))
        
        # f(x - 1) (traslación horizontal)
        f_horizontal = axes.plot(lambda x: (x-1)**2, color=BLUE, x_range=[-1, 3])
        label_h = MathTex("f(x-1)", color=BLUE, font_size=18).next_to(label_original, DOWN, aligned_edge=LEFT)
        
        info2 = Text("f(x-1): Traslación horizontal derecha", font_size=14, color=BLUE).to_edge(DOWN)
        self.play(Write(info2))
        self.play(Create(f_horizontal), Write(label_h))
        self.wait(0.5)
        
        self.play(FadeOut(f_horizontal), FadeOut(label_h), FadeOut(info2))
        
        # 2f(x) (escala vertical)
        f_escala = axes.plot(lambda x: 2*x**2, color=GREEN, x_range=[-1.4, 1.4])
        label_e = MathTex("2f(x)", color=GREEN, font_size=18).next_to(label_original, DOWN, aligned_edge=LEFT)
        
        info3 = Text("2f(x): Escala vertical (más empinada)", font_size=14, color=GREEN).to_edge(DOWN)
        self.play(Write(info3))
        self.play(Create(f_escala), Write(label_e))
        self.wait(0.5)
        
        self.play(FadeOut(f_escala), FadeOut(label_e), FadeOut(info3))
        
        # -f(x) (reflexión)
        f_reflejo = axes.plot(lambda x: -x**2, color=YELLOW, x_range=[-2, 2])
        label_r = MathTex("-f(x)", color=YELLOW, font_size=18).next_to(label_original, DOWN, aligned_edge=LEFT)
        
        info4 = Text("-f(x): Reflexión sobre el eje X", font_size=14, color=YELLOW).to_edge(DOWN)
        self.play(Write(info4))
        self.play(Create(f_reflejo), Write(label_r))
        
        self.wait()
````
